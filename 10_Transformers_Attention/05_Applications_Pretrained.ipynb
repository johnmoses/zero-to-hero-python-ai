## Introduction
Pretrained Transformer models have revolutionized AI by enabling transfer learning on large datasets, improving performance in a variety of applications.

## Core Applications

- NLP (Natural Language Processing): This is the Transformer's native domain. For translation and summarization, the classic encoder-decoder model is used to map an input text sequence to a different output text sequence. For tasks like text generation and classification, models like BERT (encoder-only) and GPT (decoder-only) have become the standard.

- Computer Vision (ViT): The Vision Transformer (ViT) reimagined how to apply Transformers to images. It breaks an image into a sequence of fixed-size patches (e.g., 16x16 pixels), flattens them, and treats them just like a sequence of word tokens in NLP. This allows the model to find complex relationships between different parts of an image, achieving state-of-the-art results that challenge the dominance of Convolutional Neural Networks (CNNs).

- Speech and Audio: Transformers are also excellent for processing audio. For speech recognition, an audio waveform is converted into a sequence of feature representations (like a spectrogram), which is then fed into a Transformer to be transcribed into a sequence of text tokens. Conversely, for text-to-speech synthesis, the model takes text as input and generates a sequence of audio features that can be converted into a realistic-sounding waveform.

## Example
Loading pretrained BERT with Hugging Face Transformers.

```python
from transformers import BertTokenizer, BertModel
import numpy as np

try:
    # Initialize tokenizer
    tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
    print("Tokenizer loaded successfully!")
    
    # Simple tokenization example
    text = "Transformers are powerful models for NLP tasks."
    tokens = tokenizer.tokenize(text)
    print(f"Tokenized text: {tokens}")
    
    # Convert tokens to IDs
    token_ids = tokenizer.convert_tokens_to_ids(tokens)
    print(f"Token IDs: {token_ids}")
    
    # Add special tokens and get attention mask
    encoded_dict = tokenizer.encode_plus(
        text,
        add_special_tokens=True,
        padding='max_length',
        max_length=20,
        truncation=True,
        return_attention_mask=True,
    )
    
    print(f"Input IDs with special tokens: {encoded_dict['input_ids']}")
    print(f"Attention mask: {encoded_dict['attention_mask']}")
    
    # Decode back to text
    decoded_text = tokenizer.decode(encoded_dict['input_ids'])
    print(f"Decoded text: {decoded_text}")

except Exception as e:
    print(f"An error occurred: {e}")
    import traceback
    traceback.print_exc()
```
