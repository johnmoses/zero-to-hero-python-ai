## Introduction
Pretrained Transformer models have revolutionized AI by enabling transfer learning on large datasets, improving performance in a variety of applications.

## Core Applications
- **NLP:** Language modeling, translation, summarization.
- **Computer Vision:** Vision Transformers (ViT).
- **Speech and Audio:** Speech recognition and synthesis.

## Example
Loading pretrained BERT with Hugging Face Transformers.

```python
from transformers import BertTokenizer, TFBertModel

# Load pretrained BERT tokenizer and model
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = TFBertModel.from_pretrained('bert-base-uncased')

# Example text
text = "Transformers are powerful models for NLP tasks."

# Tokenize input
inputs = tokenizer(text, return_tensors='tf')

# Get model outputs
outputs = model(inputs)
print(f"Last hidden state shape: {outputs.last_hidden_state.shape}")
```
