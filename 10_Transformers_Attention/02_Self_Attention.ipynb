# Self-Attention Mechanism

## Learning Objectives
- Understand the concept of self-attention in sequence models.
- Learn how self-attention computes relationships between elements in a sequence.

## Introduction
Self-attention allows models to weigh the importance of different elements in a sequence when encoding information, crucial to Transformer architectures.

## Core Concepts

- Query, Key, and Value Vectors: For each token in the input sequence, the model learns three separate vector representations:

    - Query (Q): Represents the current token's search query—what it's looking for.
    - Key (K): Represents the token's label or what it has to offer. It's what other tokens will match their queries against.
    - Value (V): Represents the actual content or information the token carries. This is what will be passed on if the token is attended to.

- Attention Scores: To determine how much focus a "query" token should place on another "key" token, a similarity score is calculated between them (typically using a dot product). This is done for the current token's query against every other token's key in the sequence. A high score means the key is highly relevant to the query.

- Softmax: The raw attention scores are then passed through a softmax function. This converts the scores into a probability distribution, resulting in a set of "attention weights" that all sum to 1. A token with a high relevance score gets a high weight, while an irrelevant token gets a weight near zero.

- Weighted Sum: Finally, the attention weights are used to create a weighted sum of all the Value vectors in the sequence. Tokens with higher attention weights contribute more of their "value" to the final output for the current position. The result is a new vector for the current token that is enriched with contextual information from all the other tokens it just paid attention to.

## Example
Manual computation of self-attention scores.

```python
import numpy as np

# Sample input matrix (3 tokens, embedding size 4)
X = np.array([
    [1, 0, 1, 0],
    [0, 2, 0, 2],
    [1, 1, 1, 1]
])

# Define Query, Key, Value matrices
W_q = np.eye(4)
W_k = np.eye(4)
W_v = np.eye(4)

Q = X.dot(W_q)
K = X.dot(W_k)
V = X.dot(W_v)

# Compute attention scores
scores = Q.dot(K.T) / np.sqrt(Q.shape[1])
print("Attention Scores:\n", scores)

# Softmax
exp_scores = np.exp(scores - np.max(scores, axis=1, keepdims=True))
attention_weights = exp_scores / exp_scores.sum(axis=1, keepdims=True)
print("\nAttention Weights:\n", attention_weights)

# Weighted sum output
output = attention_weights.dot(V)
print("\nSelf-Attention Output:\n", output)
```
