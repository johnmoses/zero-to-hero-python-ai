## Learning Objectives

## Introduction
Reinforcement learning focuses on training agents to take actions that maximize cumulative reward through interaction with an environment.

## Core Concepts
- **Agent:** Learner or decision maker.
- **Environment:** The world the agent interacts with.
- **State:** Current situation of the agent.
- **Action:** Choice made by the agent.
- **Reward:** Feedback signal evaluating actions.
- **Policy:** Strategy the agent follows.
- **Value Function:** Expected reward for states or actions.

## Example
Simple environment interaction loop pseudocode.

```python
# Pseudocode for RL agent-environment interaction
class SimpleEnv:
    def __init__(self):
        self.state = 0
    def step(self, action):
        self.state += action
        reward = -abs(self.state)
        done = abs(self.state) > 10
        return self.state, reward, done
    def reset(self):
        self.state = 0
        return self.state

env = SimpleEnv()
state = env.reset()
done = False
while not done:
    action = 1  # always move right
    state, reward, done = env.step(action)
    print(f"State: {state}, Reward: {reward}, Done: {done}")
```
