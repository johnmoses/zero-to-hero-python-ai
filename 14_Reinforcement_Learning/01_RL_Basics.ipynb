## Introduction
Reinforcement learning focuses on training agents to take actions that maximize cumulative reward through interaction with an environment.

## Core Concepts

- Agent: The AI player itself. It's the "brain" that learns and makes decisions.

- Environment: The video game world, including the rules, the level layout, and any non-player characters.

- State: A complete description of the current situation in the game from the agent's perspective. This could be the raw pixels on the screen or more structured information like the player's health, position, and the location of enemies.

- Action: A move the agent can make, such as "move left," "move right," or "jump." The set of all possible moves is the action space.

- Reward: A numerical feedback signal the agent receives from the environment after taking an action. For example, the agent might get +10 points for collecting a coin, -50 for taking damage, and +1000 for finishing the level. The agent's goal is to maximize the total cumulative reward.

- Policy: This is the agent's strategy or "brain." It's a function that takes the current state as input and decides which action to take next. The goal of RL is to find an optimal policy that maximizes the expected future reward.

- Value Function: This function estimates the long-term value of being in a particular state (or taking an action in a state). It's a prediction of the total reward the agent can expect to accumulate in the future starting from that state. The value function is often learned to help improve the policy; the agent learns that states with high values are good to be in and will try to take actions that lead to them.

## Example

```python
import numpy as np

class SimpleEnv:
    def __init__(self):
        self.state = 0
    
    def step(self, action):
        self.state += action
        reward = -abs(self.state) 
        done = abs(self.state) > 10
        return self.state, reward, done
    
    def reset(self):
        self.state = 0
        return self.state

# Create environment instance  
env = SimpleEnv()

# Reset environment
state = env.reset()
total_reward = 0
done = False

while not done:
    action = 1  # always move right
    state, reward, done = env.step(action)
    total_reward += reward

    # action = np.random.choice([-1, 1])
    # state, reward, done = env.step(action)
    # total_reward += reward

    print(f"State: {state}, Reward: {reward}, Done: {done}")

print(f"\nTotal reward: {total_reward}")
```
