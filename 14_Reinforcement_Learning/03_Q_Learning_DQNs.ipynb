# Q-Learning and Deep Q-Networks (DQNs)

## Learning Objectives
- Understand traditional Q-Learning and its limitations.
- Learn how Deep Q-Networks use neural networks to approximate Q-values.

## Introduction
Q-Learning uses a table to learn optimal actions. DQNs use neural networks to scale to large or continuous state spaces.

## Core Concepts

- Q-table: In traditional Q-learning, for environments with a small, finite number of states and actions, you can use a simple lookup table (the Q-table) to store the value of each state-action pair. However, this is completely impractical for any problem with a large or continuous state space (like images), as the table would become astronomically large.

- Function Approximation: This is the core idea of DQN. Instead of a Q-table, a deep neural network is used to approximate the Q-function. The network takes the state (e.g., the pixels of a game screen) as input and outputs the estimated Q-values for every possible action. This allows the model to handle massive state spaces and generalize to states it has never seen before.

- Experience Replay: Neural networks learn best from independent and identically distributed data, but consecutive frames in a game are highly correlated. To solve this, DQN uses an experience replay buffer. It stores the agent's experiences (state, action, reward, next_state) in a large memory. During training, it samples random mini-batches from this buffer to update the network. This breaks the temporal correlations, increases data efficiency, and smooths the learning process.

- Target Network: In Q-learning, the network's update target is calculated using its own, constantly changing weights, which can lead to instability (it's like chasing a moving target). DQN introduces a second, separate target network. This network is a periodic copy of the main Q-network and is used to generate the target values for training updates. By keeping the target network's weights fixed for a period of time, it provides a stable and consistent target for the main network to learn towards, which dramatically improves training stability.

```python
def dqn_update(batch, q_network, target_network, optimizer, gamma=0.99):
    # Unpack the batch of experiences
    states, actions, rewards, next_states, dones = batch
    
    # Convert to PyTorch tensors if needed
    states = torch.FloatTensor(states)
    actions = torch.LongTensor(actions)
    rewards = torch.FloatTensor(rewards)
    next_states = torch.FloatTensor(next_states)
    dones = torch.FloatTensor(dones)
    
    # Get current Q-values for all actions
    current_q_values = q_network(states)
    
    # Get Q-values for actions that were taken
    current_q_values = current_q_values.gather(1, actions.unsqueeze(1)).squeeze(1)
    
    # Compute target Q-values
    with torch.no_grad():
        next_q_values = target_network(next_states).max(1)[0]
        target_q_values = rewards + gamma * next_q_values * (1 - dones)
    
    # Compute loss
    loss = F.mse_loss(current_q_values, target_q_values)
    
    # Update parameters
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    return loss.item()
```
