# Q-Learning and Deep Q-Networks (DQNs)

## Learning Objectives
- Understand traditional Q-Learning and its limitations.

## Introduction
Q-Learning uses a table to learn optimal actions. DQNs use neural networks to scale to large or continuous state spaces.

## Core Concepts
- **Q-table:** Stores action values for discrete states.
- **Function Approximation:** Neural network estimates Q-values.
- **Experience Replay:** Mini-batch training from stored experiences.
- **Target Network:** Stabilizes learning by fixed parameters during updates.

## Example
DQN training step pseudocode.

```markdown
# Pseudocode for a DQN update step
def dqn_update(batch, q_network, target_network, optimizer, gamma):
    states, actions, rewards, next_states, dones = batch
    q_values = q_network(states)
    next_q_values = target_network(next_states).max(1)[0]
    targets = rewards + gamma * next_q_values * (1 - dones)
    loss = ((q_values[range(len(actions)), actions] - targets) ** 2).mean()
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    return loss.item()
```
