# Policy and Value-based Methods in RL

## Learning Objectives
- Understand policy-based and value-based RL methods.

## Introduction
Policy methods learn the action probabilities directly; value methods learn expected returns to guide decisions.

## Core Concepts

- Value Function (V-function): The state-value function, V(s), represents the expected future return (cumulative reward) if the agent starts in state s and follows its policy from then on. It essentially tells you how "good" a particular state is.

- Q-Function (Action-Value Function): The Q-function, Q(s, a), is more specific. It represents the expected future return if the agent starts in state s, takes action a, and then follows its policy. It tells you how "good" it is to take a specific action from a specific state. Q-functions are very useful because if you have the optimal Q-function, you can find the optimal policy by simply picking the action with the highest Q-value in any given state.

- Policy Gradient: This is a policy-based method. Instead of learning a value function, it directly adjusts the parameters of the agent's policy. It calculates an estimate of how much the expected reward would change if the policy parameters were changed, and then updates the policy in the direction that increases the reward (using gradient ascent). It directly learns "what to do."

- Q-Learning: This is a value-based method. Its goal is to learn the optimal Q-function. It is considered an off-policy algorithm because it can learn about the optimal policy even while executing a different, more exploratory policy. For example, it can learn the value of the best possible move in a state, even if it's currently taking random moves to explore the environment. Once the optimal Q-function is learned, the policy becomes simple: always choose the action with the highest Q-value. It learns "how good actions are" first, then derives a policy from that.

## Example
Q-learning update rule pseudocode.

```python
# Q-learning update
def q_learning_update(Q, state, action, reward, next_state, alpha, gamma):
    max_next = max(Q[next_state])
    Q[state][action] = Q[state][action] + alpha * (reward + gamma * max_next - Q[state][action])
    return Q

# Example Q-table
Q = {0: [0, 0], 1: [0, 0]}
state = 0
action = 1
reward = 1
next_state = 1
alpha = 0.1
gamma = 0.9

Q = q_learning_update(Q, state, action, reward, next_state, alpha, gamma)
print(Q)
```
