## Introduction
Neural networks are inspired by biological neurons. The perceptron was an early model that paved the way for modern deep learning.

## Core Concepts

- Perceptron: This is the simplest form of a neural network, consisting of a single neuron. It takes several binary inputs, processes them, and returns a single binary output. As a linear classifier, it can only learn patterns that are linearly separable (i.e., can be separated by a single straight line or plane).

- Weights and Bias: These are the parameters the model learns during training.
    - Weights: Each input feature is multiplied by a weight. The magnitude of the weight signifies the importance of that input in determining the output.
    - Bias: A separate, constant term that is added to the weighted sum of inputs. The bias allows the model to shift the decision boundary up or down, making it possible to find a good fit for the data even if it's not centered around the origin.

- Activation Function: This function determines the final output of the neuron. In a simple perceptron, this is typically a step function: if the weighted sum of inputs plus the bias is above a certain threshold, the neuron "fires" (outputs 1); otherwise, it outputs 0. In more complex networks, non-linear activation functions (like ReLU or Sigmoid) are used to enable the model to learn complex, non-linear patterns.

- Training: This is the process of learning the optimal values for the weights and bias. The perceptron algorithm works as follows: for each training example, it makes a prediction. If the prediction is incorrect, it adjusts the weights and bias slightly in the direction that would help it make the correct prediction for that example. This process is repeated iteratively over the entire dataset until the model's performance is satisfactory.

Basic example

```python
import numpy as np

def perceptron(x, w, b):
    z = np.dot(x, w) + b
    return 1 if z > 0 else 0

# Sample input and parameters
x = np.array([1, 2])
w = np.array([0.5, -0.6])
b = -0.1

output = perceptron(x, w, b)
print(f"Perceptron output: {output}")
```

Basic components of a perceptron

```python
import numpy as np

import matplotlib.pyplot as plt

# Demonstrate Perceptron Components
def plot_perceptron_components():
    # Create sample data
    x = np.linspace(-5, 5, 100)
    
    # Plot linear function (before activation)
    z = 2*x - 1  # example weight=2, bias=-1
    plt.figure(figsize=(12, 8))
    
    # Plot 1: Linear Function
    plt.subplot(2, 2, 1)
    plt.plot(x, z, 'b-', label='z = wx + b')
    plt.axhline(y=0, color='k', linestyle='--')
    plt.title('Linear Function\n(wx + b)')
    plt.legend()
    plt.grid(True)
    
    # Plot 2: Step Activation Function
    plt.subplot(2, 2, 2)
    y = np.where(z > 0, 1, 0)
    plt.plot(x, y, 'r-', label='Step Function')
    plt.title('Step Activation Function\n(Binary Classification)')
    plt.legend()
    plt.grid(True)
    
    # Plot 3: Weight Effect
    plt.subplot(2, 2, 3)
    weights = [0.5, 1, 2]
    for w in weights:
        plt.plot(x, w*x - 1, label=f'w={w}')
    plt.title('Effect of Weights\n(Different Slopes)')
    plt.legend()
    plt.grid(True)
    
    # Plot 4: Bias Effect
    plt.subplot(2, 2, 4)
    biases = [-2, 0, 2]
    for b in biases:
        plt.plot(x, 2*x + b, label=f'b={b}')
    plt.title('Effect of Bias\n(Vertical Shift)')
    plt.legend()
    plt.grid(True)
    
    plt.tight_layout()
    plt.show()

# Call the function to visualize
plot_perceptron_components()

# Print explanation
print("Perceptron Components Explained:")
print("1. Linear Function: Combines inputs with weights (wx + b)")
print("2. Step Function: Converts continuous output to binary decision")
print("3. Weights: Control the slope/importance of each input")
print("4. Bias: Shifts the decision boundary")
```
