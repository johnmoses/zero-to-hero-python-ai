## Introduction
Neural networks are inspired by biological neurons. The perceptron was an early model that paved the way for modern deep learning.

## Core Concepts
- **Perceptron:** Simple linear binary classifier.
- **Weights and Bias:** Model parameters.
- **Activation Function:** Introduces non-linearity.
- **Training:** Adjusting weights based on errors.


```python
import numpy as np

def perceptron(x, w, b):
    z = np.dot(x, w) + b
    return 1 if z > 0 else 0

# Sample input and parameters
x = np.array([1, 2])
w = np.array([0.5, -0.6])
b = -0.1

output = perceptron(x, w, b)
print(f"Perceptron output: {output}")
```
