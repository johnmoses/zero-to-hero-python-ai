## Introduction
This notebook covers popular classical algorithms for unsupervised learning including clustering and dimensionality reduction.

## Core Concepts
- **k-Means:** Partition data into k clusters by minimizing distance to cluster centers.
- **DBSCAN:** Density-based clustering identifying core and noise points.
- **PCA:** Linear dimensionality reduction capturing maximum variance.

## Example
Applying k-Means and PCA.

```python
import numpy as np
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, DBSCAN

# Create sample data
X = np.random.rand(200, 4)

# k-Means clustering with 3 clusters
kmeans = KMeans(n_clusters=3, random_state=42)
labels_km = kmeans.fit_predict(X)

# DBSCAN clustering
dbscan = DBSCAN(eps=0.1, min_samples=5)
labels_db = dbscan.fit_predict(X)

# PCA to 2D
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

print(f"k-Means cluster labels (first 10): {labels_km[:10]}")
print(f"DBSCAN cluster labels (first 10): {labels_db[:10]}")
```
