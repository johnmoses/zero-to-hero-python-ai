## Introduction
This notebook covers popular classical algorithms for unsupervised learning including clustering and dimensionality reduction.

## Core Concepts

- k-Means: This is a popular clustering algorithm that aims to partition a dataset into a pre-determined number of clusters, 'k'. It works iteratively: 1) It randomly initializes 'k' cluster centers (centroids). 2) It assigns each data point to its nearest centroid. 3) It updates each centroid to be the mean of all data points assigned to it. Steps 2 and 3 are repeated until the cluster assignments no longer change. The result is a set of compact, non-overlapping spherical clusters.

- DBSCAN (Density-Based Spatial Clustering of Applications with Noise): Unlike k-Means, DBSCAN does not require the number of clusters to be specified beforehand. It groups together points that are closely packed in high-density areas, and it can discover clusters of arbitrary shape. It works by identifying "core points" (points with many nearby neighbors) and expanding clusters from them. Points that are not reachable from any core points are labeled as "noise" or outliers.

- PCA (Principal Component Analysis): This is the most common technique for dimensionality reduction. It transforms a dataset with many correlated features into a smaller set of new, uncorrelated features called "principal components." These components are ordered by the amount of original data variance they capture. By keeping only the first few principal components, you can reduce the complexity of your data while retaining most of its important information, which is useful for visualization and improving model performance.

## Example
Applying k-Means and PCA.

```python
import numpy as np
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, DBSCAN

# Create sample data
X = np.random.rand(200, 4)

# k-Means clustering with 3 clusters
kmeans = KMeans(n_clusters=3, random_state=42)
labels_km = kmeans.fit_predict(X)

# DBSCAN clustering
dbscan = DBSCAN(eps=0.1, min_samples=5)
labels_db = dbscan.fit_predict(X)

# PCA to 2D
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

print(f"k-Means cluster labels (first 10): {labels_km[:10]}")
print(f"DBSCAN cluster labels (first 10): {labels_db[:10]}")
```
