## Introduction
Supervised learning is a core type of machine learning where models are trained using labeled data. The model learns to map inputs to known outputs to make predictions on new data.

## Core Concepts
- **Training Data:** Dataset with input features and labeled outputs.
- **Classification:** Predicting discrete labels.
- **Regression:** Predicting continuous values.
- **Model Training:** Learning the mapping from inputs to outputs.
- **Evaluation:** Measuring performance using metrics like accuracy or mean squared error.

## Workflow
1. Collect and preprocess data
2. Select and train model
3. Evaluate model
4. Tune hyperparameters
5. Deploy model

```python
# Example: Simple classification problem
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=42)

knn = KNeighborsClassifier()
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"KNN Accuracy: {accuracy:.2f}")
```
