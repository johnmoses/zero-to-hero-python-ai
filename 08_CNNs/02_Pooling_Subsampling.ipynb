# Pooling and Subsampling

## Learning Objectives
- Understand pooling operations and their role in CNNs.
- Learn max pooling and average pooling.

## Introduction
Pooling reduces the spatial size of feature maps, decreasing computation and helping network become invariant to small translations.

## Core Concepts

- Max Pooling: This is a downsampling operation that reduces the dimensionality of the feature map, which helps to decrease computational complexity and control overfitting. By selecting the most activated feature in a patch, it preserves the most prominent features detected by the convolution layer.

- Average Pooling: Like max pooling, this operation also downsamples the feature map. However, instead of taking the most prominent feature, it provides a more generalized and smoother representation of the features within the window. It's less common than max pooling but can be useful in certain architectures.

- Stride and Pooling Size:

    - Pooling Size (or Window/Kernel Size): Defines the N x M block of the feature map from which to pool values. A 2x2 window is common.
    - Stride: This is the step size the window takes as it moves across the feature map. A stride equal to the pooling size (e.g., a 2x2 window with a stride of 2) results in non-overlapping windows and is a very common way to downsample the feature map by a factor of the stride.

## Example
Manual max pooling on a feature map.

```python
import numpy as np

feature_map = np.array([
    [1, 3, 2, 4],
    [5, 6, 7, 8],
    [3, 2, 1, 0],
    [1, 2, 3, 4]
])

def max_pooling(feature_map, size=2, stride=2):
    output_dim = (feature_map.shape[0] - size) // stride + 1
    pooled = np.zeros((output_dim, output_dim))
    for i in range(output_dim):
        for j in range(output_dim):
            patch = feature_map[i*stride:i*stride+size, j*stride:j*stride+size]
            pooled[i, j] = np.max(patch)
    return pooled

pooled_map = max_pooling(feature_map)
print("Pooled Feature Map (Max Pooling):")
print(pooled_map)
```
