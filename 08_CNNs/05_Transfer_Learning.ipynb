## Introduction
Transfer learning leverages knowledge from pretrained models on large datasets to solve similar tasks with less data and faster training.

## Core Concepts

- Pretrained Models: These models serve as a powerful starting point for new tasks. Since they have already learned a rich hierarchy of features (e.g., edges, textures, patterns, and object parts) from a massive dataset, they provide a significant advantage over training a model from scratch, especially when you have limited data for your specific task.

- Fine-tuning: This is a common transfer learning technique. You start with a pretrained model, freeze the early layers (which learned general features), and unfreeze the later, more task-specific layers. You then continue training these unfrozen layers (and often a new classifier head) on your own dataset, typically with a very low learning rate. This adapts the high-level features of the pretrained model to the specifics of your new task.

- Feature Extraction: This is another, simpler transfer learning technique. Here, you use the convolutional base of a pretrained model as a fixed feature extractor. You pass your new data through the network and use the output of one of the intermediate layers as a set of features. These features are then fed into a new, separate classifier (e.g., a simple neural network or an SVM) that you train from scratch. Unlike fine-tuning, the weights of the pretrained model are not updated.

## Example
Using pretrained MobileNetV2 for a new classification task.

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Load pretrained MobileNetV2 without top layer
base_model = tf.keras.applications.MobileNetV2(input_shape=(224, 224, 3),
                                               include_top=False,
                                               weights='imagenet')
base_model.trainable = False  # freeze base

# Add new classification head
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(5, activation='softmax')  # example 5 classes
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.summary()
```
