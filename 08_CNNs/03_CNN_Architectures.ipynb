## Introduction
CNN architectures have evolved to increase depth, improve feature extraction, and address training issues via structural innovations.

## Core Architectures

- LeNet-5: One of the earliest successful CNNs, it established the standard architecture of stacking convolutional and pooling layers, followed by fully connected layers for classification. Its success on the MNIST dataset helped popularize CNNs for image classification tasks.

- AlexNet: Its victory in the 2012 ImageNet competition was a pivotal moment for deep learning. It was much deeper and wider than previous networks and was the first to effectively use the ReLU activation function (which helped with training speed) and dropout for regularization. Its success demonstrated the power of deep CNNs trained on large datasets with GPU acceleration.

- VGG: The key innovation of VGGNet was its simplicity and uniformity. It demonstrated that a stack of small 3x3 convolutional filters could achieve the same effective receptive field as larger filters (e.g., 5x5 or 7x7) but with more non-linearities, making the decision function more discriminative and the network easier to train. This showed the value of depth in network design.

- ResNet (Residual Network): This architecture solved the "degradation" problem, where adding more layers to a network would hurt its performance. The "skip connections" allow the network to learn an identity function, meaning a layer can be skipped if it's not useful. This enables the training of networks that are hundreds or even thousands of layers deep, which was previously impossible due to vanishing gradients.

## Example
Summary of ResNet block in PyTorch pseudocode.

```python
# PyTorch example for a basic residual block
import torch.nn as nn

class ResidualBlock(nn.Module):
    def __init__(self, channels):
        super(ResidualBlock, self).__init__()
        self.conv1 = nn.Conv2d(channels, channels, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(channels)
        self.relu = nn.ReLU(inplace=True)
        self.conv2 = nn.Conv2d(channels, channels, kernel_size=3, padding=1)
        self.bn2 = nn.BatchNorm2d(channels)

    def forward(self, x):
        identity = x
        out = self.conv1(x)
        out = self.bn1(out)
        out = self.relu(out)
        out = self.conv2(out)
        out = self.bn2(out)
        out += identity
        out = self.relu(out)
        return out

print("Residual Block defined.")
```
