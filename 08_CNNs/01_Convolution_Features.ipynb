# Convolution and Feature Extraction

## Learning Objectives
- Understand the convolution operation and its role in CNNs.
- Learn how convolutions extract local features from images.

## Introduction
Convolution layers apply filters to input images to extract important features like edges, textures, and shapes. This forms the fundamental building block of convolutional neural networks.

## Core Concepts
- **Convolution:** Sliding a filter (kernel) over input to produce feature maps.
- **Filters:** Small matrices that detect specific patterns.
- **Stride and Padding:** Control the movement of filters and output size.
- **Feature Maps:** Outputs of convolution layers capturing detected features.

## Example
Manual convolution operation on a small image patch.

```python
import numpy as np

# Sample 5x5 grayscale image
image = np.array([
    [1, 2, 0, 2, 1],
    [3, 1, 2, 0, 0],
    [0, 1, 3, 1, 2],
    [2, 0, 1, 3, 1],
    [1, 2, 0, 0, 2]
])

# Sample 3x3 filter (edge detection)
filter = np.array([
    [1, 0, -1],
    [1, 0, -1],
    [1, 0, -1]
])

def convolve2d(image, filter):
    filter_size = filter.shape[0]
    output_dim = image.shape - filter_size + 1
    output = np.zeros((output_dim, output_dim))
    for i in range(output_dim):
        for j in range(output_dim):
            patch = image[i:i+filter_size, j:j+filter_size]
            output[i, j] = np.sum(patch * filter)
    return output

feature_map = convolve2d(image, filter)
print("Feature Map:")
print(feature_map)
```
