## Learning Objectives

## Introduction
Video data offers temporal information useful in applications such as surveillance, activity recognition, and autonomous driving.

## Core Concepts
- **Frame extraction:** Processing individual video frames.
- **Motion detection:** Identifying moving objects.
- **Object tracking:** Following objects across frames.

## Example
Extract frames from a video file using OpenCV.

```python
import cv2

# Video file path
video_path = 'path_to_video.mp4'  # Replace with your video file

# Capture video
cap = cv2.VideoCapture(video_path)

frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret or frame_count >= 10:  # Read first 10 frames
        break
    cv2.imshow('Frame', frame)
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break
    frame_count += 1

cap.release()
cv2.destroyAllWindows()
print(f"Displayed {frame_count} frames.")
```
