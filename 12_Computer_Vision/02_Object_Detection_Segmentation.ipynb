## Introduction
Object detection locates objects in images while segmentation assigns labels to each pixel, enabling detailed scene understanding.

## Core Concepts

- Bounding Boxes: This is the primary way to specify an object's location. A bounding box is a rectangle, typically represented by the coordinates of its corners (e.g., top-left and bottom-right), that encloses an object. The goal of an object detection model is to output a class label (e.g., "cat") and the coordinates of a bounding box for each object it finds.

- Segmentation Masks: While a bounding box provides a coarse location, a segmentation mask is much more precise. It is a pixel-level map where every pixel belonging to an object is assigned a specific class. This provides the exact shape and contour of the object, not just a rectangular approximation. This is the output of a "semantic segmentation" or "instance segmentation" model.

- Anchor Boxes: Many modern object detection models (like YOLO and Faster R-CNN) don't search for objects blindly. Instead, they use a predefined set of default bounding boxes of various sizes and aspect ratios, known as anchor boxes. The model's task is then to predict how to shift and scale these anchor boxes to better fit the actual objects in the image and to assign a class label to them. This approach makes the detection process much more efficient.

- Evaluation: Evaluating object detection requires measuring both localization and classification accuracy.

    - Intersection over Union (IoU): This metric quantifies how well a predicted bounding box overlaps with the ground-truth box. It's calculated as the area of their overlap divided by the area of their union. An IoU of 1 means a perfect match. A prediction is typically considered correct if the IoU is above a set threshold (e.g., 0.5).
    
    - mAP (mean Average Precision): This is the gold standard metric for object detection. It calculates the Average Precision (AP) for each class—a metric that summarizes the precision/recall curve—and then takes the mean of these AP values across all classes. It provides a single, comprehensive score that reflects the model's performance across different confidence levels and classes.

## Example
Using a pretrained Mask R-CNN model for segmentation.

```python
# Requires torchvision and matplotlib
import torch
import torchvision.transforms as T
from PIL import Image
import matplotlib.pyplot as plt
import requests
from io import BytesIO
from torchvision.models.detection import maskrcnn_resnet50_fpn

# Load image
url = 'https://images.unsplash.com/photo-1572276596237-6847af57a37c'
response = requests.get(url)
img = Image.open(BytesIO(response.content)).convert("RGB")

# Preprocess
transform = T.Compose([T.ToTensor()])
img_tensor = transform(img)

# Load pretrained model
model = maskrcnn_resnet50_fpn(pretrained=True)
model.eval()

# Predict
with torch.no_grad():
    prediction = model([img_tensor])[0]

print("Detected objects:")
print(prediction['labels'])
```
