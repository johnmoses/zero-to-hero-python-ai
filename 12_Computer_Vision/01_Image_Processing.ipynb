## Introduction
Computer Vision in the field of AI enables computer systems to deduce meaningful interpretations from images or any other visual stimulation and take action based on that input. Computer vision gives the system the ability to observe.

Pattern recognition is the foundation of today's computer vision algorithms. We train computers on a vast amount of visual data—images are processed, things are labeled, and patterns are found in those items.

Image processing prepares images for analysis by enhancing features, reducing noise, or extracting useful information.

## Core Concepts

- Image Transformations: These are geometric operations that change the orientation or size of an image.

    - Resizing is used to make images a consistent size.
    - Cropping is used to select a specific region of interest.
    - Rotating changes the image's orientation. These are very common preprocessing steps and are also used for data augmentation to help train models more robustly.

- Filtering: This involves applying a kernel (a small matrix) across an image to modify pixel values based on their neighbors. This can enhance or suppress certain features.

    - Blurring (e.g., with a Gaussian filter) smooths the image by averaging pixel values, which is useful for reducing noise.
    - Sharpening accentuates the differences between pixels, making edges and details appear crisper.

- Feature Detection: This involves using algorithms to find interesting and salient points or regions in an image. Instead of looking at all pixels, models can focus on these key features.

    - Edge detection (e.g., using the Canny algorithm) identifies pixels where there is a sharp change in intensity, outlining objects.
    - Corner detection finds points where two edges intersect, which are often stable and useful features for tracking or matching objects.

- Color Space Conversion: Digital images can be represented in different color models (spaces). Converting between them can make certain tasks easier.

    - RGB to Grayscale: This is the most common conversion. It simplifies the image from three color channels to one intensity channel, reducing complexity for algorithms that don't need color information.
    - RGB to HSV (Hue, Saturation, Value): This space is often more intuitive as it separates color (Hue) from its intensity (Value). This is very useful for tasks like color-based object tracking, as the hue is less affected by changes in lighting.

## Example
Basic image processing with OpenCV.

```python
import cv2
import matplotlib.pyplot as plt

# Load image
img = cv2.imread('https://upload.wikimedia.org/wikipedia/commons/7/7d/Dog_face.png')

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Resize image
resized = cv2.resize(gray, (100, 100))

# Display
plt.imshow(resized, cmap='gray')
plt.title('Grayscale Resized Image')
plt.axis('off')
plt.show()
```
