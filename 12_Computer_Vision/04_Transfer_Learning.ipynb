# Transfer Learning in Computer Vision

## Learning Objectives
- Learn how to apply pretrained models to new CV tasks.
- Understand feature extraction and fine-tuning approaches.

## Introduction
Transfer learning leverages pretrained networks trained on large datasets to improve performance and reduce training time on related tasks.

## Core Concepts
- **Feature extraction:** Use pretrained model as fixed feature extractor.
- **Fine-tuning:** Retrain some upper layers on new data.
- **Common pretrained models:** ResNet, VGG, MobileNet.

## Example
Using pretrained ResNet for feature extraction with PyTorch.

```python
import torch
import torchvision.models as models
from torch import nn

# Load pretrained ResNet50
resnet50 = models.resnet50(pretrained=True)

# Freeze all layers
for param in resnet50.parameters():
    param.requires_grad = False

# Replace classification layer
num_classes = 10
resnet50.fc = nn.Linear(resnet50.fc.in_features, num_classes)

print(resnet50)
```
