## Introduction
Transfer learning leverages pretrained networks trained on large datasets to improve performance and reduce training time on related tasks.

## Core Concepts

- Feature Extraction: This is one of the two main strategies for transfer learning. In this approach, you take a powerful model that has been pretrained on a very large dataset (like ImageNet) and treat it as a fixed feature extractor. You remove the final classification part of the network and pass your new images through the frozen convolutional base. The output is a vector of high-level features for each image, which you can then use to train a new, much smaller classifier for your specific task. This is very efficient as you only have to train the new classifier.

- Fine-tuning: This is a more powerful strategy. Instead of freezing the entire convolutional base, you unfreeze the top few layers and continue to train them on your new dataset (typically with a very low learning rate). The idea is that the early layers of the network learned very general features (like edges and colors), which should be kept, while the later layers learned more specific features that can be "fine-tuned" to your new task. This usually results in better performance than feature extraction but requires more data.

- Common Pretrained Models: These are well-known, high-performing architectures for which pretrained weights (usually from the ImageNet dataset) are widely available.

    - ResNet and VGG are classic, powerful models known for their accuracy.
    - MobileNet is a family of lightweight, efficient models designed specifically for on-device and mobile applications where computational resources are limited. Using one of these as a starting point for transfer learning is standard practice and saves an enormous amount of time and data compared to training a model from scratch.

## Example
Using pretrained ResNet for feature extraction with PyTorch.

```python
import torch
import torchvision.models as models
from torch import nn

# Load pretrained ResNet50
resnet50 = models.resnet50(pretrained=True)

# Freeze all layers
for param in resnet50.parameters():
    param.requires_grad = False

# Replace classification layer
num_classes = 10
resnet50.fc = nn.Linear(resnet50.fc.in_features, num_classes)

print(resnet50)
```
