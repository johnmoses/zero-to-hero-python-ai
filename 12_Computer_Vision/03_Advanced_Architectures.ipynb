## Introduction
Recent architectures improve model accuracy and efficiency by introducing skip connections, dense connections, and computational optimizations.

## Core Architectures

- ResNet (Residual Network): ResNet's key innovation is the "skip connection" or "residual block." As networks get deeper, they become harder to train due to the vanishing gradient problem. A skip connection creates a shortcut that allows the input of a block of layers to be added directly to its output. This makes it easier for gradients to flow through the network and allows the model to easily learn "identity functions" (i.e., just pass the input through unchanged), which in turn makes it feasible to train extremely deep and powerful networks (over 100 layers) without performance degradation.

- DenseNet (Densely Connected Network): DenseNet takes the concept of feature reuse to the next level. In a DenseNet, each layer is connected to every other layer in a feed-forward fashion. This means that each layer receives the feature maps from all preceding layers as its input. This dense connectivity encourages the network to reuse features, strengthens gradient flow, and generally leads to very high parameter efficiency, often achieving top performance with fewer parameters than comparable ResNets.

- EfficientNet: The insight behind EfficientNet is that the performance of a CNN depends on a balance between its depth (number of layers), width (number of channels), and the resolution of the input image. Instead of scaling just one of these dimensions, EfficientNet proposes a "compound scaling" method that scales all three dimensions simultaneously in a principled way. This allows it to create a family of models that are extremely efficient, achieving better accuracy with far fewer parameters and computations than previous state-of-the-art models.
