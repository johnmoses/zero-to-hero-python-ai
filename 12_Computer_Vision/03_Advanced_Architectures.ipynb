## Introduction
Recent architectures improve model accuracy and efficiency by introducing skip connections, dense connections, and computational optimizations.

## Core Architectures
- **ResNet:** Uses residual connections to enable very deep networks.
- **DenseNet:** Connects each layer to every other layer to improve feature reuse.
- **EfficientNet:** Balances depth, width, and resolution through compound scaling.
