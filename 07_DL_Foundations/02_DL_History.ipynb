## Introduction
Deep learning evolved over decades from early neural network models to today's highly successful architectures. Understanding its history helps contextualize current trends.

## Core Milestones
- **1943:** McCulloch and Pitts propose first neural model.
- **1958:** Rosenblatt invents the Perceptron.
- **1980s:** Backpropagation algorithm enables deeper networks.
- **2006:** Deep belief networks reinvigorate interest.
- **2012:** AlexNet wins ImageNet with convolutional deep nets.
- **Recent:** Transformers revolutionize NLP and beyond.
