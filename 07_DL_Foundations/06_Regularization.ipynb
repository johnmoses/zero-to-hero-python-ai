# Regularization Techniques

## Learning Objectives
- Understand the role of regularization in preventing overfitting.
- Learn common regularization methods: L1, L2, Dropout.

## Introduction
Regularization helps control model complexity and improves generalization by penalizing large weights or randomly dropping units during training.

## Core Concepts
- **L1 Regularization:** Promotes sparsity by adding absolute value penalties.
- **L2 Regularization:** Adds squared penalties to reduce weight magnitude.
- **Dropout:** Randomly drops neurons during training to prevent co-adaptation.

## Example
Applying L2 regularization in TensorFlow.

```python
import tensorflow as tf
from tensorflow.keras import layers, regularizers

model = tf.keras.Sequential([
    layers.Dense(64, activation='relu', kernel_regularizer=regularizers.l2(0.01), input_shape=(10,)),
    layers.Dense(1)
])

model.compile(optimizer='adam', loss='mse')
model.summary()
```
