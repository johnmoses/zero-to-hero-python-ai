# Loss Functions and Optimization

## Learning Objectives
- Learn about common loss functions and their applications.
- Understand optimization algorithms like gradient descent.

## Introduction
Loss functions quantify prediction errors. Optimization algorithms minimize these losses to train neural networks.

## Core Concepts

- Loss Functions:

    - MSE (Mean Squared Error): Used for regression tasks (predicting a continuous value like price or temperature). It calculates the average of the squared differences between the predicted and actual values, heavily penalizing larger errors.
    
    - Cross-Entropy: Used for classification tasks. It measures how different the predicted class probabilities are from the actual class (which is a one-hot distribution). A lower cross-entropy value means the model's predictions are more confident and more accurate.

- Optimization:

    - Gradient Descent: The core process of iteratively adjusting model parameters to minimize the loss function. The "vanilla" version calculates the gradient using the entire dataset for each update, which is slow.
    
    - Variants: The most common variant is Mini-Batch Gradient Descent, which offers a balance of efficiency and stability by calculating the gradient and updating parameters based on a small, random subset (a "mini-batch") of the data at a time. Stochastic Gradient Descent (SGD) is the extreme case where the batch size is just one.

- Learning Rate: This is a critical hyperparameter that determines the size of the steps the optimizer takes during each update. If it's too small, training will be very slow; if it's too large, the optimizer may overshoot the optimal solution and fail to converge.

- Momentum, Adam: These are popular optimization algorithms that enhance basic gradient descent.

    - Momentum: Helps accelerate the optimizer in the correct direction and dampens oscillations by adding a fraction of the previous update step to the current one. It's like a ball rolling down a hill, gathering momentum.
    
    - Adam (Adaptive Moment Estimation): A highly effective and widely used optimizer that combines the idea of momentum with adaptive learning rates. It calculates a different learning rate for each parameter, making it very efficient and often a good default choice that requires less manual tuning.

## Example
Define MSE loss and perform one gradient descent step.

```python
import numpy as np

# Sample data
y_true = np.array([0.5])
y_pred = np.array([0.4])

# MSE loss and gradient
loss = np.mean((y_true - y_pred) ** 2)
grad = 2 * (y_pred - y_true) / y_true.size

# Gradient descent step
learning_rate = 0.1
y_pred_new = y_pred - learning_rate * grad

print(f"Loss: {loss}")
print(f"Updated prediction after gradient step: {y_pred_new}")
```
