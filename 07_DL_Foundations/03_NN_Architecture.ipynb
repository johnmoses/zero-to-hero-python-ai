## Introduction
Neural networks consist of layers of connected neurons which learn to represent data transformations and patterns.

## Core Concepts

- Neurons: Think of a neuron as a small computational unit. It takes multiple numbers as inputs, multiplies each by a "weight" (representing its importance), sums them up, adds a value called a "bias," and then passes this final result through an activation function to produce its output.

- Layers:

    - Input Layer: The entry point of the network, where the raw data (e.g., pixels of an image, features of a dataset) is fed in.
    
    - Hidden Layers: Intermediate layers between the input and output. This is where the network learns to identify complex patterns and features by combining the outputs from previous layers.
    
    - Output Layer: The final layer that produces the network's prediction or result (e.g., classifying an image as "cat" or "dog").

- Activation Functions: These are mathematical functions applied by neurons to their output. Their crucial role is to introduce non-linearity, which allows the network to learn patterns that are more complex than a simple straight line.

    - Sigmoid: Compresses any input into a value between 0 and 1.
    - ReLU (Rectified Linear Unit): A very common and efficient function that outputs the input if it is positive, and 0 otherwise.
    tanh (Hyperbolic Tangent): Compresses any input into a value between -1 and 1.
    - Feedforward: This describes the directional flow of information in the network. Data moves in one direction only—from the input layer, through the hidden layers, to the output layer—without looping back.

- Parameters: These are the internal variables that the network "learns" from the data during the training process.

    - Weights: Control the strength of the connection between two neurons.
    - Biases: A value that allows a neuron to shift its activation function, providing more flexibility to fit the data.

## Example
Simple feedforward calculation.

```python
import numpy as np

# Simple neuron with ReLU activation
def relu(x):
    return np.maximum(0, x)

x = np.array([1.0, 2.0])
w = np.array([[0.5, -0.6], [0.1, 0.8]])
b = np.array([0.1, -0.2])

# Calculate layer output
z = np.dot(x, w) + b
a = relu(z)
print(f"Layer output: {a}")
```
