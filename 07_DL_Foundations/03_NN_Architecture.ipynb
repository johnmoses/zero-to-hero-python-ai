## Introduction
Neural networks consist of layers of connected neurons which learn to represent data transformations and patterns.

## Core Concepts
- **Neurons:** Basic units computing weighted sums followed by activation.
- **Layers:** Input, hidden, and output layers.
- **Activation Functions:** Sigmoid, ReLU, tanh introduce non-linearity.
- **Feedforward:** Data flow from input to output.
- **Parameters:** Weights and biases learned during training.

## Example
Simple feedforward calculation.

```python
import numpy as np

# Simple neuron with ReLU activation
def relu(x):
    return np.maximum(0, x)

x = np.array([1.0, 2.0])
w = np.array([[0.5, -0.6], [0.1, 0.8]])
b = np.array([0.1, -0.2])

# Calculate layer output
z = np.dot(x, w) + b
a = relu(z)
print(f"Layer output: {a}")
```
