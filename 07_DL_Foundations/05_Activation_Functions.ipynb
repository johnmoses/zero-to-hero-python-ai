# Activation Functions

## Learning Objectives
- Learn different activation functions used in neural networks.
- Understand their properties and when to use them.

## Introduction
Activation functions introduce non-linearity to neural networks, enabling them to learn complex patterns.

## Core Concepts

- Sigmoid: This function squashes any input value into a range between 0 and 1, which is useful for output layers where a probability is needed (e.g., binary classification). Its "smooth gradient" means it's differentiable everywhere, but it can lead to the "vanishing gradient" problem in deep networks, where gradients become too small to effectively update weights.

- Tanh (Hyperbolic Tangent): This function squashes any input value into a range between -1 and 1. Being "zero-centered" means its outputs are, on average, closer to zero, which can help speed up convergence during training compared to Sigmoid. It's often preferred over Sigmoid for hidden layers.

- ReLU (Rectified Linear Unit): This function is very simple: it outputs the input directly if it's positive, and outputs zero otherwise. This makes it extremely "efficient" to compute. It also promotes "sparse activation" because many neurons will output zero for a given input, which can make the network more efficient and easier to optimize. It is the most common default choice for hidden layers.

- Leaky ReLU: This is a variant of ReLU that, instead of outputting zero for negative inputs, outputs a very small positive value (e.g., 0.01 times the input). This "leak" ensures the neuron never completely "dies"—the "dying ReLU problem" occurs when a neuron gets stuck outputting zero and can no longer update its weights via gradient descent. Leaky ReLU ensures there is always a small gradient, even for negative inputs.

## Example
Plot and calculate activations.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.1):
    return np.where(x > 0, x, alpha * x)

plt.plot(x, sigmoid(x), label='Sigmoid')
plt.plot(x, tanh(x), label='Tanh')
plt.plot(x, relu(x), label='ReLU')
plt.plot(x, leaky_relu(x), label='Leaky ReLU')
plt.legend()
plt.title('Activation Functions')
plt.show()
```
