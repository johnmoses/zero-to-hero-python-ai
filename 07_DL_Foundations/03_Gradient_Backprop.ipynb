# Gradient and Backpropagation

## Learning Objectives

## Introduction
Backpropagation calculates gradients of the loss function with respect to model parameters, enabling optimization via gradient descent.

## Core Concepts
- **Gradient:** Direction and rate of fastest increase of function.
- **Chain Rule:** Used to compute gradients of complex, composed functions.
- **Loss Function:** Measures error between predictions and true values.
- **Gradient Descent:** Updates weights to minimize loss.

## Simple Example
Backpropagation for a single neuron.

```python
import numpy as np

# Inputs and weights
x = np.array([1.0, 2.0])
w = np.array([0.5, -0.5])
b = 0.0

# True label
y_true = 1

# Forward pass
z = np.dot(x, w) + b
y_pred = 1 / (1 + np.exp(-z))  # Sigmoid

# Loss (binary cross-entropy)
loss = -(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))

# Backward pass (gradients)
dL_dy = -(y_true / y_pred) + ((1 - y_true) / (1 - y_pred))
dy_dz = y_pred * (1 - y_pred)
dz_dw = x

grad_w = dL_dy * dy_dz * dz_dw
grad_b = dL_dy * dy_dz

print(f"Loss: {loss}")
print(f"Gradient w.r.t weights: {grad_w}")
print(f"Gradient w.r.t bias: {grad_b}")
```
