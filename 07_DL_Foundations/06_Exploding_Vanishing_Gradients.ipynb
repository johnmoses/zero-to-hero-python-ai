# Exploding and Vanishing Gradients

## Learning Objectives
- Understand the exploding and vanishing gradient problems in deep networks.
- Learn strategies to mitigate these issues.

## Introduction
During training, extremely large or small gradients can hinder learning. This notebook discusses why these problems occur and how to address them.

## Core Concepts
- **Vanishing Gradients:** Gradients become too small in deep layers, slowing or stopping learning.
- **Exploding Gradients:** Gradients become too large, causing unstable updates.
- **Mitigation Strategies:** Proper initialization, normalization, gradient clipping, residual connections.


```python
# Example: Gradient clipping in PyTorch
import torch
from torch import nn, optim

model = nn.Linear(10, 1)
optimizer = optim.SGD(model.parameters(), lr=0.1)

for input in [torch.randn(10) for _ in range(5)]:
    optimizer.zero_grad()
    output = model(input)
    loss = output.sum()
    loss.backward()
    # Clip gradients
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
    optimizer.step()

print("Gradient clipping example run complete.")
```
