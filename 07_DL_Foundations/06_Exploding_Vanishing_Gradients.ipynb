# Exploding and Vanishing Gradients

## Learning Objectives
- Understand the exploding and vanishing gradient problems in deep networks.
- Learn strategies to mitigate these issues.

## Introduction
During training, extremely large or small gradients can hinder learning. This notebook discusses why these problems occur and how to address them.

## Core Concepts

- Vanishing Gradients: This occurs in deep networks when gradients are backpropagated from the output layer to the initial layers. With certain activation functions (like Sigmoid), the gradients are repeatedly multiplied by small numbers, causing them to shrink exponentially until they are virtually zero. When the gradient is this small, the weights in the early layers do not update, and the network effectively stops learning from the data.

- Exploding Gradients: This is the opposite problem, where repeated multiplication during backpropagation causes gradients to become excessively large. When these huge gradients are used to update the network's weights, they cause massive, unstable changes. This can lead to the model's loss becoming NaN (Not a Number) and prevents the network from converging to a good solution.

- Mitigation Strategies:

    - Proper Initialization: Using smart weight initialization schemes (like He or Xavier initialization) sets the initial weights to values that are not too large or too small, which helps prevent gradients from immediately vanishing or exploding at the start of training.

    - Normalization (e.g., Batch Norm): This technique normalizes the inputs to each layer to have a stable distribution (e.g., zero mean, unit variance). This helps keep the gradients in a reasonable range throughout the network.
    
    - Gradient Clipping: This is a direct solution for exploding gradients. It involves setting a predefined threshold for the maximum gradient value. If a gradient exceeds this threshold during an update, it is scaled down, preventing the update from being too large.
    
    - Residual Connections (ResNets): This architectural innovation introduces "skip connections" that allow the gradient to bypass one or more layers. This creates a shortcut for the gradient to flow back through the network, which is highly effective at combating the vanishing gradient problem in very deep models.


```python
# Example: Gradient clipping in PyTorch
import torch
from torch import nn, optim

model = nn.Linear(10, 1)
optimizer = optim.SGD(model.parameters(), lr=0.1)

for input in [torch.randn(10) for _ in range(5)]:
    optimizer.zero_grad()
    output = model(input)
    loss = output.sum()
    loss.backward()
    # Clip gradients
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
    optimizer.step()

print("Gradient clipping example run complete.")
```
