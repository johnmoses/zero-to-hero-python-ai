## Introduction
Artificial Intelligence (AI) is an interdisciplinary field of computer science focused on building systems capable of performing tasks that typically require human intelligence. These tasks include recognizing speech, understanding natural language, making decisions, visual perception, and problem solving.

The origin of AI dates back to the 1950s, inspired by the idea that anything the human mind can do, a computer can be made to simulate. Since then, AI has evolved significantly—from symbolic AI and expert systems to modern machine learning and deep learning approaches that learn from data.

AI today powers many technologies we use daily, from digital assistants like Siri and Alexa, to recommendation systems on Netflix, to self-driving cars, and advanced diagnostic tools in healthcare.

### Types of AI
- **Narrow AI:** AI systems designed to perform a specific task, such as image recognition or playing chess.
- **General AI:** Hypothetical AI that can perform any intellectual task a human can do.
- **Supervised, Unsupervised, and Reinforcement Learning:** Key learning paradigms featured later in this course.

This course will focus on practical approaches within Narrow AI, emphasizing data-driven learning systems.

## Core Concepts

AI systems exhibit several fundamental capabilities:

- **Learning from Data:** AI algorithms use patterns in data to improve their performance without being explicitly programmed for every rule. This capability is the core of machine learning.

- **Reasoning and Problem Solving:** AI systems can apply logical rules or learned knowledge to make decisions or solve complex problems, such as recommendation engines or game-playing AI.

- **Perception:** The ability to interpret sensory data like images, sounds, and text. For example, computer vision enables AI to recognize objects in images.

- **Natural Language Understanding:** Allows computers to comprehend and generate human language, powering chatbots, translation systems, and voice assistants.

- **Automation:** AI systems automate repetitive or dangerous tasks, improving efficiency and safety across various industries.

### AI System Components
- **Data:** Raw inputs from the real world, essential for training AI models.
- **Algorithms:** Procedures or formulas enabling the AI to learn and make predictions.
- **Models:** The mathematical representation—trained from data—that AI uses to perform tasks.
- **Compute Resources:** Hardware such as CPUs and GPUs enabling large-scale data processing.

### Classifications of AI Systems
- **Rule-Based Systems:** Early AI, operating with hand-coded rules.
- **Machine Learning:** Systems learn from data rather than rules; focus of modern AI.
- **Deep Learning:** Subset of ML using neural networks with many layers, powering state-of-the-art applications.

In upcoming chapters, we will unpack each of these concepts in detail, building the foundations to develop and deploy real AI models.
