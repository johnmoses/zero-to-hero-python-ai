## Learning Objectives

## Introduction
We use the MNIST dataset, a popular benchmark in AI, consisting of 28x28 pixel images of handwritten digits (0-9). This notebook guides you step-by-step through creating a simple feedforward neural network for classification.

## Core Concepts
### Dataset Loading
- MNIST dataset contains 70,000 grayscale images (60,000 training, 10,000 test)
- Each image is 28x28 pixels, representing handwritten digits 0-9
- Built-in function `tf.keras.datasets.mnist.load_data()` handles the download and splitting

### Normalization
- Raw pixel values range from 0 (black) to 255 (white)
- Division by 255 scales values to [0, 1] range
- Benefits:
    - Prevents numerical instability during training
    - Helps gradient descent converge faster
    - Makes training more stable

### Model Architecture
1. **Flatten Layer**: Converts 28x28 images into 784-element vectors
2. **Dense Layer (128 units)**:
     - Fully connected layer with ReLU activation
     - Learns complex patterns in the data
3. **Dropout (0.2)**:
     - Randomly deactivates 20% of neurons during training
     - Prevents overfitting by adding regularization
4. **Output Layer (10 units)**:
     - One unit per digit class (0-9)
     - Softmax activation outputs probability distribution

### Loss Function
- Sparse Categorical Crossentropy:
    - Standard for multi-class classification
    - Measures difference between predicted and actual digit
    - "Sparse" means labels are integers (0-9) rather than one-hot encoded

### Optimizer
- Adam (Adaptive Moment Estimation):
    - Combines benefits of RMSprop and momentum
    - Adapts learning rates for each parameter
    - Generally provides good results without much tuning

### Evaluation
- Model tested on 10,000 unseen images
- Accuracy metric shows percentage of correct predictions
- Test set performance indicates real-world generalization

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize pixel values
x_train, x_test = x_train / 255.0, x_test / 255.0

# Build model
model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(10, activation='softmax')
])

# Compile model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Train model
model.fit(x_train, y_train, epochs=5)

# Evaluate
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f'Test accuracy: {test_acc:.4f}')
```
