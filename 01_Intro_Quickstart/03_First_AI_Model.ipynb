# Your First AI Model: MNIST Digit Classification

## Learning Objectives
- Build a basic neural network using TensorFlow to classify handwritten digits.
- Understand key steps: data loading, normalization, model construction, training, and evaluation.

## Introduction
We use the MNIST dataset, a popular benchmark in AI, consisting of 28x28 pixel images of handwritten digits (0-9). This notebook guides you step-by-step through creating a simple feedforward neural network for classification.

## Core Concepts
- **Dataset Loading:** Fetching and preparing the MNIST dataset.
- **Normalization:** Scaling pixel values to the range [0, 1] for numerical stability.
- **Model Architecture:** A sequence of layers including input flattening, dense layers, dropout, and output softmax.
- **Loss Function:** Sparse categorical crossentropy for multi-class classification.
- **Optimizer:** Adam algorithm for adaptive learning.
- **Evaluation:** Measuring accuracy on unseen test data.

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize pixel values
x_train, x_test = x_train / 255.0, x_test / 255.0

# Build model
model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(10, activation='softmax')
])

# Compile model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Train model
model.fit(x_train, y_train, epochs=5)

# Evaluate
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f'Test accuracy: {test_acc:.4f}')
```
