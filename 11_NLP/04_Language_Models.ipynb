## Introduction
Language models predict the probability of sequences of words, essential for NLP tasks like speech recognition and text generation.

## Core Concepts

- N-gram Models: This is the classic statistical approach. An "n-gram" is a sequence of 'n' words. The model is based on the assumption that the probability of a word depends only on the n-1 words that came before it. For example, in a trigram (n=3) model, to predict the next word after "the cat," you would look at all the trigrams in your training data that start with "the cat" and count how often each word follows. While effective to a degree, they fail to capture long-distance dependencies and struggle with word sequences they've never seen before (sparsity).

- Neural Language Models: These are modern language models that use neural networks (like RNNs or Transformers) to predict the next word. Instead of relying on counts, they learn continuous vector representations (embeddings) of words. This allows them to understand semantic similarity (e.g., knowing "cat" and "dog" are similar) and capture much more complex, long-range patterns in language. This ability to generalize is why they have become far more powerful than n-gram models.

- Perplexity: This is the primary metric for evaluating a language model. It measures how well a probability model predicts a sample. A lower perplexity score indicates that the model is less "surprised" by the words in a test set, meaning it assigns a higher probability to the sequences that actually occurred. A perplexity of 100, for example, can be loosely interpreted as the model being as confused as if it had to choose uniformly among 100 different words at each step. Lower is better.

## Example
Train a simple n-gram model.

```python
from collections import Counter

text = "the quick brown fox jumps over the lazy dog"
tokens = text.split()

bigrams = [(tokens[i], tokens[i+1]) for i in range(len(tokens)-1)]

bigram_counts = Counter(bigrams)
unigram_counts = Counter(tokens[:-1])

def bigram_prob(w1, w2):
    return bigram_counts[(w1, w2)] / unigram_counts[w1]

print(f"P(jumps | fox) = {bigram_prob('fox', 'jumps')}")
```
