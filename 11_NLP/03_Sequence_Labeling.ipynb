## Introduction
Sequence labeling assigns labels to each element in a sequence, a fundamental task in NLP applications like syntactic parsing.

## Core Concepts

- Conditional Random Fields (CRFs): A CRF is a type of probabilistic model that is exceptionally good at sequence labeling. Instead of labeling each token independently, a CRF considers the labels of neighboring tokens. This allows it to learn dependencies and constraints between labels (e.g., an "I-PERSON" tag representing the inside of a person's name must follow a "B-PERSON" tag representing the beginning). This global perspective on the entire sequence's labels often leads to more coherent and accurate predictions than making isolated decisions for each token.

- Recurrent Models (LSTMs and GRUs): These models are often used to provide rich, context-aware features to a CRF. A common and powerful architecture is the BiLSTM-CRF. In this setup, a bidirectional LSTM processes the input text to generate a representation for each word that includes context from both before and after it. These sophisticated representations are then used as inputs to the CRF layer, which makes the final labeling decision. The LSTM captures the semantic context, while the CRF ensures the final sequence of labels is structured correctly.

- Evaluation Metrics: For sequence labeling tasks, simple accuracy is often misleading. These metrics provide a more nuanced view of performance:

    - Precision: Answers the question: "Of all the entities I predicted, how many were actually correct?" It measures the model's exactness.
    
    - Recall: Answers the question: "Of all the entities that actually exist in the data, how many did my model find?" It measures the model's completeness.
    
    - F1-Score: This is the harmonic mean of precision and recall (2 * (Precision * Recall) / (Precision + Recall)). It provides a single, balanced score and is often the primary metric for comparing models, as it punishes systems that excel at one metric while sacrificing the other.

## Example
POS tagging using simple rule-based or pretrained model.

```python
import nltk

# First, make sure we download all required resources
try:
    # Download the punkt tokenizer
    nltk.download('punkt')
    
    # Download the POS tagger
    nltk.download('averaged_perceptron_tagger')
    
    # Now proceed with tokenization and tagging
    sentence = "Natural language processing is fascinating."
    
    # Tokenize the sentence
    tokens = nltk.word_tokenize(sentence)
    print(f"Tokens: {tokens}")
    
    # Now perform POS tagging
    tags = nltk.pos_tag(tokens)
    print(f"POS Tags: {tags}")
    
except Exception as e:
    print(f"An error occurred: {str(e)}")
    import traceback
    traceback.print_exc()
```
