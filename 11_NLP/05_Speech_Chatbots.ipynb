## Introduction
Speech and chatbots combine NLP and audio processing to enable natural human-computer interactions.

## Core Concepts
- **Speech Recognition:** Converting audio to text (ASR).
- **Speech Synthesis:** Text-to-speech generation.
- **Chatbots:** Rule-based and AI-driven conversational agents.
- **Dialogue Management:** Keeping context in conversations.

## Example
Use SpeechRecognition library for audio input.

```python
# Note: Requires microphone and audio packages
# !pip install SpeechRecognition pyaudio
import speech_recognition as sr

r = sr.Recognizer()
with sr.Microphone() as source:
    print("Say something...")
    audio = r.listen(source)

try:
    text = r.recognize_google(audio)
    print(f"You said: {text}")
except Exception as e:
    print(f"Error: {e}")
```
