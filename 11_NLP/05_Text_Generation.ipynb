## Introduction
Text generation involves producing natural language sequences, widely used in chatbots, translation, and creative writing.

## Core Concepts
- **Language Models:** Probabilistic generation of next tokens.
- **Sampling Techniques:** Greedy, beam search, temperature sampling.
- **Evaluation:** Perplexity, human evaluation.

## Example
Generate text with a simple RNN model.

```python
import tensorflow as tf
import numpy as np

# Dummy char-to-index mapping and reverse
char2idx = {c:i for i, c in enumerate('abcdefghijklmnopqrstuvwxyz ')}
idx2char = np.array(list(char2idx.keys()))

# Sample input sequence
seq = 'hello world'
input_eval = [char2idx[c] for c in seq]
input_eval = tf.expand_dims(input_eval, 0)

# Simple RNN model
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(len(char2idx), 64),
    tf.keras.layers.SimpleRNN(128, return_sequences=True, stateful=True),
    tf.keras.layers.Dense(len(char2idx))
])

# Generate logits and sample next char
predictions = model(input_eval)
pred_logits = predictions[:, -1, :]
predicted_id = tf.random.categorical(pred_logits, num_samples=1)[-1, 0].numpy()
print(f"Next predicted char: {idx2char[predicted_id]}")
```
