## Introduction
Text generation involves producing natural language sequences, widely used in chatbots, translation, and creative writing.

## Core Concepts

- Language Models (for Generation): At their core, language models are next-token predictors. When generating text, they take an existing sequence of tokens (the prompt) and output a probability distribution over their entire vocabulary for what the next token should be. The process of selecting a token, adding it to the sequence, and repeating is how text is generated.

- Sampling Techniques: How you select a token from that probability distribution is crucial and defines the output's characteristics.

    - Greedy Search: The simplest method. At each step, it picks the single token with the highest probability. It's fast but often leads to repetitive and predictable text.
    
    - Beam Search: An improvement over greedy search. It keeps track of the k most probable sequences (the "beams") at each step and ultimately chooses the sequence with the highest overall probability. It generally produces more fluent text but can still lack creativity.
    
    - Temperature Sampling: A method to control randomness. It reshapes the probability distribution. A high "temperature" flattens the distribution, making less likely words more probable and leading to more creative or even nonsensical output. A low temperature makes the distribution sharper, behaving more like greedy search.

- Evaluation: Measuring the quality of generated text is notoriously difficult.

    - Perplexity: This is an intrinsic, automated metric that measures how well a model predicts a given text. While useful for comparing models during training, a low perplexity doesn't always guarantee high-quality, coherent, or interesting generated text.
    
    - Human Evaluation: This is considered the gold standard. It involves having humans rate the generated text on subjective qualities like fluency, coherence, relevance, and creativity. It's slow and expensive but provides the most meaningful assessment of a model's real-world performance.

## Example
Generate text with a simple RNN model.

```python
import tensorflow as tf
import numpy as np

# Dummy char-to-index mapping and reverse
char2idx = {c:i for i, c in enumerate('abcdefghijklmnopqrstuvwxyz ')}
idx2char = np.array(list(char2idx.keys()))

# Sample input sequence
seq = 'hello world'
input_eval = [char2idx[c] for c in seq]
input_eval = tf.expand_dims(input_eval, 0)

# Simple RNN model
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(len(char2idx), 64),
    tf.keras.layers.SimpleRNN(128, return_sequences=True, stateful=True),
    tf.keras.layers.Dense(len(char2idx))
])

# Generate logits and sample next char
predictions = model(input_eval)
pred_logits = predictions[:, -1, :]
predicted_id = tf.random.categorical(pred_logits, num_samples=1)[-1, 0].numpy()
print(f"Next predicted char: {idx2char[predicted_id]}")
```
