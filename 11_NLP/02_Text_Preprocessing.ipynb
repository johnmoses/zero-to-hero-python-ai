## Introduction
Text preprocessing is critical to convert raw text into structured input data for NLP models.

## Core Concepts

- Tokenization: This is the foundational step of breaking down a body of text into smaller, manageable units called tokens. Most commonly, tokens are words, but they can also be characters or sub-word units depending on the specific needs of the task.

- Stopword Removal: This is the process of filtering out common words that provide little semantic value to a sentence, such as "the," "a," "is," and "in." By removing them, you allow the model to focus on the more important words that define the text's meaning.

- Stemming: This is a fast, rule-based process for reducing words to their root form, or "stem." It typically works by chopping off prefixes and suffixes. The result is not guaranteed to be a real dictionary word. For example, "studies," "studying," and "student" might all be stemmed to the root "studi."

- Lemmatization: This is a more advanced and linguistically informed method of reducing words to their dictionary form, known as the "lemma." Unlike stemming, it considers the word's part of speech and uses a vocabulary to ensure the output is a valid word. For example, the word "better" would be lemmatized to "good," and "ran" would become "run." It is more computationally expensive than stemming but more accurate.

## Example
Preprocess example sentence.

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize

# nltk.download('punkt')
# nltk.download('stopwords')
# nltk.download('wordnet')

text = "Natural language processing enables computers to understand human language."
tokens = word_tokenize(text.lower())

stop_words = set(stopwords.words('english'))
filtered_tokens = [w for w in tokens if w.isalpha() and w not in stop_words]

stemmer = PorterStemmer()
stemmed = [stemmer.stem(w) for w in filtered_tokens]

lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in filtered_tokens]

print("Tokens:", tokens)
print("Filtered Tokens:", filtered_tokens)
print("Stemmed:", stemmed)
print("Lemmatized:", lemmatized)
```
