# Text Preprocessing in NLP

## Learning Objectives
- Understand common text preprocessing techniques.
- Learn tokenization, stopword removal, stemming, and lemmatization.

## Introduction
Text preprocessing is critical to convert raw text into structured input data for NLP models.

## Core Concepts
- **Tokenization:** Splitting text into meaningful units.
- **Stopword Removal:** Removing common words that carry little meaning.
- **Stemming:** Reducing words to their root forms.
- **Lemmatization:** Reducing words to base dictionary form.

## Example
Preprocess example sentence.

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

text = "Natural language processing enables computers to understand human language."
tokens = word_tokenize(text.lower())

stop_words = set(stopwords.words('english'))
filtered_tokens = [w for w in tokens if w.isalpha() and w not in stop_words]

stemmer = PorterStemmer()
stemmed = [stemmer.stem(w) for w in filtered_tokens]

lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in filtered_tokens]

print("Tokens:", tokens)
print("Filtered Tokens:", filtered_tokens)
print("Stemmed:", stemmed)
print("Lemmatized:", lemmatized)
```
