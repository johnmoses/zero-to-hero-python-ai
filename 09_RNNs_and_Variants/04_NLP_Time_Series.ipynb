## Introduction
RNNs are widely used in NLP tasks like language modeling, text generation, and in forecasting time series data due to their sequential memory capabilities.

## Core Concepts

- Tokenization: This is the initial step of breaking raw text into a sequence of smaller units called "tokens." These tokens can be as simple as words separated by spaces, or they can be sub-word units (e.g., "tokenization" becomes "token" and "##ization"). This process converts a string of text into a structured sequence that a model can ingest one piece at a time.

- Embedding: Since models work with numbers, not text, each token is mapped to a numerical vector through an embedding layer. This isn't just a simple lookup; the embedding vector is a learned representation that captures the token's semantic meaning. In a well-trained model, words with similar meanings will have similar embedding vectors, allowing the model to understand relationships like "king" is to "queen" as "man" is to "woman."

- Sequence-to-Sequence (Seq2Seq) Modeling: This is a powerful framework for tasks where the input is a sequence and the output is also a sequence, and their lengths may differ. It typically uses two RNNs (or Transformers): an Encoder that processes the input sequence and compresses its meaning into a single context vector (the final hidden state), and a Decoder that takes that context vector and generates the output sequence token by token. This architecture is the foundation for applications like machine translation, text summarization, and conversational bots.

## Example
Basic text vectorization and passing sequences through an RNN layer.

```python
import tensorflow as tf
from tensorflow.keras.layers import Embedding, SimpleRNN
from tensorflow.keras.models import Sequential

# Vocabulary size and embedding dimension
vocab_size = 1000
embedding_dim = 16

# Sample input sequence (batch_size=2, sequence_length=5)
input_sequences = tf.constant([[1, 2, 3, 4, 5], [5, 4, 3, 2, 1]])

# Build model
model = Sequential([
    Embedding(vocab_size, embedding_dim, input_length=5),
    SimpleRNN(32)
])

output = model(input_sequences)
print(f"RNN output shape: {output.shape}")
```
