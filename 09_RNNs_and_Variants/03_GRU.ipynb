# Gated Recurrent Unit (GRU)

## Learning Objectives
- Learn about the GRU architecture and gating mechanisms.
- Understand GRU advantages over LSTM.

## Introduction
GRUs simplify LSTM design with fewer gates, offering faster computation while maintaining performance on many tasks.

## Core Concepts

- Update Gate: This single gate handles the roles of both the forget and input gates in an LSTM. It decides how much of the previous hidden state to keep and how much of the new, candidate hidden state to incorporate. It essentially determines the balance between old and new information.

- Reset Gate: This gate controls how much of the previous hidden state is used to compute the new candidate hidden state. When the reset gate is close to 0, it effectively makes the unit act as if it's processing the first element of a sequence, allowing it to forget its past context and focus on the new input.

- Simpler than LSTM, fewer parameters: GRUs have a less complex architecture than LSTMs. They merge the cell state and hidden state into a single state and use only two gates instead of three. This means they have fewer weights (parameters) to train, which can make them computationally more efficient and slightly faster, while often delivering performance comparable to LSTMs.

## Example
GRU forward pass calculations.

```python
import numpy as np
from scipy.special import expit

def gru_cell(x, h_prev, Wz, Wr, Wh, bz, br, bh):
    z = expit(np.dot(Wz, x) + bz)  # update gate
    r = expit(np.dot(Wr, x) + br)  # reset gate
    h_tilde = np.tanh(np.dot(Wh, x) + r * h_prev + bh)  # candidate hidden state
    h = (1 - z) * h_prev + z * h_tilde
    return h

# Input and initial hidden state
x = np.array([0.5, 0.1])
h_prev = np.array([0.0, 0.0])

# Random weights and biases
Wz = np.random.rand(2, 2)
Wr = np.random.rand(2, 2)
Wh = np.random.rand(2, 2)
bz = np.random.rand(2)
br = np.random.rand(2)
bh = np.random.rand(2)

h = gru_cell(x, h_prev, Wz, Wr, Wh, bz, br, bh)
print("GRU hidden state:", h)
```
