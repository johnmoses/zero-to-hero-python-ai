# Long Short-Term Memory (LSTM)

## Learning Objectives
- Understand the LSTM architecture and its components.
- Learn how LSTMs solve vanilla RNN limitations.

## Introduction
LSTMs incorporate gating mechanisms to retain and forget information, effectively addressing the vanishing gradient problem.

## Core Concepts

- Cell State The cell state is the most critical element of the LSTM. It acts as a long-term memory or an "information conveyor belt" that runs straight down the entire processing chain. Information can flow along it with very few linear transformations, making it easy for dependencies from early in a sequence to be carried to the end. This is the primary mechanism LSTMs use to combat the vanishing gradient problem. The gates (forget, input, output) are responsible for carefully regulating what information is added to or removed from this cell state.

- Forget Gate This gate's job is to decide which information to discard from the cell state. It looks at the previous hidden state (the short-term memory) and the current input, and for each number in the previous cell state, it outputs a value between 0 and 1. A '1' represents "completely keep this information," while a '0' represents "completely get rid of this." This allows the LSTM to selectively forget things that are no longer relevant. For example, if it's processing a story, it might forget the gender of the old protagonist when a new one is introduced.

- Input Gate The input gate is responsible for deciding what new information to add to the cell state. This process has two steps:

    - First, a sigmoid layer (the "input gate") decides which parts of the new information are worth paying attention to (outputting values between 0 and 1).
    
    - Second, a tanh layer creates a vector of all possible new candidate values that could be added. Finally, the output of the input gate is multiplied with the candidate values to scale down the information, ensuring only the important parts are added to the cell state. This prevents irrelevant inputs from polluting the long-term memory.

- Output Gate The output gate determines the next hidden state, which is the "short-term memory" used for making predictions at the current time step. This hidden state is a filtered version of the cell state. The gate first runs a sigmoid layer on the previous hidden state and current input to decide which parts of the cell state should be output. Then, the current cell state is passed through a tanh function to squash its values between -1 and 1. These two results are multiplied together to produce the new hidden state. This allows the LSTM to expose only the relevant parts of its long-term memory for the task at hand.

## Example
Simplified LSTM cell forward computation.

```python
import numpy as np
from scipy.special import expit  # sigmoid function

def lstm_cell(x, h_prev, c_prev, Wf, Wi, Wo, Wc, bf, bi, bo, bc):
    # Forget gate
    f = expit(np.dot(Wf, x) + bf)
    # Input gate
    i = expit(np.dot(Wi, x) + bi)
    # Candidate cell state
    c_tilde = np.tanh(np.dot(Wc, x) + bc)
    # New cell state
    c = f * c_prev + i * c_tilde
    # Output gate
    o = expit(np.dot(Wo, x) + bo)
    # New hidden state
    h = o * np.tanh(c)
    return h, c

# Input and previous states
x = np.array([0.5, 0.1])
h_prev = np.array([0.0, 0.0])
c_prev = np.array([0.0, 0.0])

# Random weights and biases
Wf = np.random.rand(2, 2)
Wi = np.random.rand(2, 2)
Wo = np.random.rand(2, 2)
Wc = np.random.rand(2, 2)
bf = np.random.rand(2)
bi = np.random.rand(2)
bo = np.random.rand(2)
bc = np.random.rand(2)

h, c = lstm_cell(x, h_prev, c_prev, Wf, Wi, Wo, Wc, bf, bi, bo, bc)
print("Hidden state:", h)
print("Cell state:", c)
```
