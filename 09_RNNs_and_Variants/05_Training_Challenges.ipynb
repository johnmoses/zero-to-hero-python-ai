# Training Challenges with RNNs

## Learning Objectives
- Understand the challenges of training RNNs such as vanishing and exploding gradients.
- Learn strategies to mitigate these issues.

## Introduction
RNNs face difficulties in learning long-range dependencies due to vanishing/exploding gradients. Various techniques help address these problems.

## Core Challenges
- **Vanishing gradients:** Gradients shrink exponentially.
- **Exploding gradients:** Gradients grow exponentially and destabilize learning.
- **Training instability:** Difficulty in convergence.

## Mitigation Techniques
- Gradient clipping.
- Use LSTM or GRU cells.
- Careful initialization and normalization.

## Example
Applying gradient clipping in TensorFlow.

```python
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Embedding(1000, 64),
    tf.keras.layers.LSTM(64),
    tf.keras.layers.Dense(1)
])

optimizer = tf.keras.optimizers.Adam(learning_rate=0.001, clipnorm=1.0)
model.compile(optimizer=optimizer, loss='mse')

print('Model compiled with gradient clipping.')
```
