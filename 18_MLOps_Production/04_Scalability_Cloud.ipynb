## Introduction
Cloud platforms provide resources to scale ML models for training and serving, managing fluctuating workloads efficiently.

## Core Concepts

- Horizontal Scaling: Also known as scaling out, this is the strategy of increasing a system's capacity by adding more machines (or nodes) to a resource pool. Instead of making a single server more powerful (which is vertical scaling or scaling up), you distribute the workload across multiple servers. For a machine learning model deployed as an API, this would mean running the same model API on several servers behind a load balancer. As traffic increases, the load balancer directs requests to the different servers, preventing any single one from being overwhelmed and ensuring consistent response times.

- Autoscaling: This is an intelligent, dynamic form of horizontal scaling where the number of active servers is automatically adjusted based on real-time demand. You define rules based on metrics like CPU utilization or the number of incoming requests. For example, you could set a rule to add a new server whenever the average CPU usage across your servers exceeds 70% for five minutes. Conversely, you can set a rule to remove servers when traffic is low (e.g., CPU usage drops below 30%), which helps in reducing costs by not paying for idle resources. It's a core feature of modern cloud computing that provides both high availability and cost-efficiency.

- Cloud Providers: These are companies that offer a wide range of computing services over the internet. For machine learning, the "big three" are:

    - AWS (Amazon Web Services): Offers a mature and extensive suite of services. Its flagship ML platform, Amazon SageMaker, provides an end-to-end solution for building, training, and deploying models, along with other AI services for vision, speech, and more.
    
    - GCP (Google Cloud Platform): Known for its strength in data analytics, big data, and AI. Its Vertex AI platform unifies its ML offerings, providing tools for everything from data labeling to model deployment and monitoring. Google also has a strong focus on custom hardware like TPUs for training large models.
    
    - Azure: Microsoft's cloud platform has strong enterprise integration. Azure Machine Learning is its comprehensive service that supports the entire machine learning lifecycle. It integrates well with other Microsoft products and offers features like a visual "designer" for building models with low-code interfaces.

- Serverless and Containerized Deployment: These are two modern approaches to deploying applications that offer great flexibility and efficiency:

- Containerized Deployment: This involves packaging an application's code, libraries, and all its dependencies into a single, isolated unit called a container (Docker is the most popular tool for this). This container can then run on any machine that has the containerization software, ensuring that the application works consistently everywhere, from a developer's laptop to a production server. This solves the classic "it works on my machine" problem.

- Serverless Deployment: This is a cloud computing model where the cloud provider dynamically manages the allocation and provisioning of servers. You simply upload your code (often as a "function"), and the provider runs it in response to triggers (like an API request). You don't have to worry about the underlying infrastructure at all, and you only pay for the actual compute time your function uses. AWS Lambda and Google Cloud Functions are popular examples.

These two concepts are often used together. For example, you can run a container on a serverless platform like AWS Fargate, getting the portability of containers without having to manage the servers.
