# Bias Mitigation in AI

## Learning Objectives
- Understand sources of bias in AI datasets and models.
- Learn common techniques to detect and mitigate bias.

## Introduction
Bias in AI systems leads to unfair outcomes and must be actively identified and addressed.

## Sources of Bias

- Dataset Imbalance or Selection Bias: This is one of the most common sources of bias. It occurs when the data used to train a model does not accurately represent the real-world environment where it will be used.

    - Imbalance: A classic example is a facial recognition system trained on a dataset where 80% of the images are of people from one ethnicity. The system will likely be less accurate for underrepresented ethnicities.

    - Selection Bias: This happens when the data is not collected randomly. For instance, if training data for a loan-approval model is sourced only from a bank's existing, affluent customers, the model may unfairly deny loans to applicants from different economic backgrounds, even if they are creditworthy.

- Algorithmic Bias: This refers to bias that is introduced by the model's algorithm itself. While algorithms are just math, their design can have unintended consequences. For example, an algorithm designed to maximize ad clicks might inadvertently learn to target high-paying job ads predominantly to one gender if historical data shows a correlation, thus reinforcing existing societal inequalities rather than promoting equal opportunity.

- Human Biases in Labeling: Most supervised machine learning models are trained on data that has been labeled by humans. These human labelers can unintentionally project their own conscious or unconscious biases onto the data. For example, if human reviewers label resumes for a "good candidate" model, their inherent biases about what makes a good candidate (e.g., favoring graduates from certain universities) will be learned and automated by the model.

## Mitigation Strategies

- Preprocessing: Data Balancing, Augmentation. This category of strategies focuses on fixing the data before it is used to train the model.

    - Data Balancing: If a dataset is imbalanced, techniques like over-sampling the minority class (creating copies of existing data) or under-sampling the majority class (removing data) can be used to create a more balanced distribution.

    - Augmentation: This involves creating new, synthetic data points to enrich underrepresented categories. For image data, this could mean flipping, rotating, or changing the lighting on existing images to create more training examples.

- In-processing: Fairness Constraints in Training. This approach modifies the model's training process itself to reduce bias. Instead of just optimizing for accuracy, the algorithm is also given a "fairness metric" to optimize for. For example, you could add a constraint that requires the model's prediction error rate to be roughly equal across different demographic groups. This forces the model to learn in a way that is not only accurate but also equitable.

- Post-processing: Adjusting Model Outputs. This strategy involves adjusting the model's predictions after the model has already been trained. For example, if a model that predicts credit risk is found to have a higher threshold for a certain group, a post-processing step could be applied to calibrate the thresholds for different groups to ensure the final decisions are fair. This is often easier to implement than the other two methods but can be less effective as it doesn't address the root cause of the bias within the model itself.


```python
import numpy as np
from sklearn.utils import resample

# Example: balancing dataset classes
def balance_data(X, y):
    classes = np.unique(y)
    max_count = max([sum(y == c) for c in classes])
    X_balanced, y_balanced = [], []
    for c in classes:
        X_c = X[y == c]
        y_c = y[y == c]
        X_resampled, y_resampled = resample(X_c, y_c, replace=True, n_samples=max_count)
        X_balanced.append(X_resampled)
        y_balanced.append(y_resampled)
    return np.vstack(X_balanced), np.hstack(y_balanced)
```
