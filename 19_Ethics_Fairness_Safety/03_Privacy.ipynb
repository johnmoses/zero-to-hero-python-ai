## Introduction
Protecting individual's data privacy is crucial for ethical and legal AI deployments.

## Privacy Risks

- Data Leaks from Model Inversion: This is a sophisticated attack where an adversary uses the outputs of a trained model to reconstruct the private data it was trained on. By repeatedly querying the model and analyzing its predictions (especially the confidence scores), an attacker can reverse-engineer and piece together sensitive information about individuals in the original dataset. For example, a facial recognition model could be "inverted" to generate images of the specific faces it learned from, even if the attacker never had access to the original photos.

- Unauthorized Data Access: This is a more traditional security threat, but it's amplified for AI systems. AI often requires centralizing vast amounts of data for training, creating a single, high-value target for attackers. A breach could expose not just the raw training data (e.g., patient medical records, customer financial details), but also the trained model itself, which is a valuable asset that contains learned patterns from that sensitive data.

## Privacy-Enhancing Techniques (PETs)

- Differential Privacy: This is a formal mathematical framework for providing privacy guarantees. The core idea is to add a carefully calibrated amount of statistical "noise" to the data or to the results of queries on the data. This noise is significant enough to mask the contribution of any single individual, making it nearly impossible for an attacker to determine if a specific person's data was part of the dataset. This allows organizations to share aggregate insights from data while protecting the privacy of the individuals within it. There is, however, a fundamental trade-off: more noise provides more privacy but can reduce the accuracy of the model.

- Federated Learning: This is a decentralized training technique that flips the traditional machine learning paradigm on its head. Instead of bringing all the raw data to a central server to train a model, the model is sent out to be trained on the devices where the data resides (e.g., individual mobile phones or hospital servers). Each device trains a local version of the model on its own data, and only the resulting model updates (small, aggregated numerical values called gradients) are sent back to the central server. These updates are then combined to improve the shared global model. The key benefit is that the raw, sensitive data never leaves the local device, dramatically enhancing user privacy.

## Example
Basic illustration of differential privacy noise mechanism.

```python
import numpy as np

def add_dp_noise(data, epsilon=1.0):
    sensitivity = 1.0
    scale = sensitivity / epsilon
    noise = np.random.laplace(0, scale, data.shape)
    return data + noise

# Example data
data = np.array([10, 20, 30])
noisy_data = add_dp_noise(data)
print("Original ", data)
print("Noisy data with differential privacy:", noisy_data)
```
