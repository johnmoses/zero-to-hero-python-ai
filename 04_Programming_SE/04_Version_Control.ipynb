# Version Control with Git

## Learning Objectives
- Understand the basics of version control and Git.
- Learn how to initialize a repository, commit changes, and manage branches.

## Introduction
Version control systems like Git are essential for managing code changes, collaboration, and project history in AI development.

## Core Concepts
- **Repository:** A directory where your project files and history are stored.
- **Commit:** A snapshot of your changes.
- **Branch:** Parallel versions of the project.
- **Merge:** Combining branches.
- **Remote Repository:** Hosted copy of your repository (e.g., GitHub).

## Example
Basic Git commands.

```bash
# The following commands are to be run in the terminal, not in Python.
#
# git init  # Initialize a new repository
# git add .  # Stage files for commit
# git commit -m "Initial commit"  # Commit changes
# git branch feature  # Create a new branch
# git checkout feature  # Switch to feature branch
# git merge main  # Merge main branch into feature
```
