## Introduction
Version control systems like Git are essential for managing code changes, collaboration, and project history in AI development.

## Core Concepts
Repository (Repo)

Think of this as your project's main folder. It contains all your project files (code, images, etc.) and, crucially, a complete history of every change ever made to those files. This history is stored in a hidden sub-folder called .git.

Commit

A commit is a snapshot of your repository at a specific point in time. When you've made changes (like adding a feature or fixing a bug), you "commit" them. This saves your changes to the project history with a unique ID and a message describing what you did. It's like saving a checkpoint in a video game.

Branch

A branch is a parallel version of your repository. It allows you to work on new features or experimental ideas without affecting the main, stable version of your project (which is often on a branch called main or master). You can create a new branch, make commits on it, and it won't change the main branch until you're ready.

Merge

Merging is the process of taking the changes from one branch and integrating them into another. For example, after you've successfully built a new feature on its own branch, you would merge that branch back into the main branch to make the feature part of the official project.

Remote Repository

This is a version of your repository that is hosted on the internet or another network, most commonly on a service like GitHub, GitLab, or Bitbucket. It acts as a central place for you and your team to share changes, collaborate on code, and back up your work. You "push" your local commits to the remote repository and "pull" changes from others.
