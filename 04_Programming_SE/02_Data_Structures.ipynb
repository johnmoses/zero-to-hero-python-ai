## Core Concepts

Lists

Think of a list as a shopping list. It's an ordered sequence of items, and you can change it (it's mutable). You can add items, remove them, or change their order. Lists can also contain duplicate items.

Example:

```python
fruits = ["apple", "banana", "cherry"]
fruits.append("orange")
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'orange']
```

Tuples

A tuple is like a list, but once you create it, you can't change it (it's immutable). This is useful for data that shouldn't be modified, like coordinates (e.g., (x, y)).

Example:

```python
point = (10, 20)
# point[0] = 15  # This would cause an error
print(point) # Output: (10, 20)
```

Dictionaries

Dictionaries store data in key-value pairs, much like a real dictionary where you look up a word (the key) to find its definition (the value). Keys must be unique.

Example:

```python
person = {"name": "Alice", "age": 30}
print(person["name"])  # Output: Alice
```

Sets

Sets are unordered collections of unique elements. If you add a duplicate item to a set, it will only be stored once. They are useful for tasks like removing duplicates from a list or checking for membership.

Example:

```python
numbers = {1, 2, 3, 2, 1}
print(numbers)  # Output: {1, 2, 3}
```

## Example

```python
# Example usage of data structures
fruits = ['apple', 'banana', 'cherry']  # list
coordinates = (10.0, 20.0)              # tuple
student_grade = {'Alice': 90, 'Bob': 85} # dictionary
unique_numbers = {1, 2, 3, 2, 1}        # set

print(f"Fruits: {fruits}")
print(f"Coordinates: {coordinates}")
print(f"Student Grades: {student_grade}")
print(f"Unique Numbers: {unique_numbers}")
```
