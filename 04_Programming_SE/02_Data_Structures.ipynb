## Introduction
AI programming often requires efficient handling of data. This notebook introduces popular Python data structures and their use cases.

## Core Concepts
- **Lists:** Ordered, mutable collections.
- **Tuples:** Ordered, immutable collections.
- **Dictionaries:** Key-value mappings.
- **Sets:** Unordered collections of unique elements.

## Example

```python
# Example usage of data structures
fruits = ['apple', 'banana', 'cherry']  # list
coordinates = (10.0, 20.0)              # tuple
student_grade = {'Alice': 90, 'Bob': 85} # dictionary
unique_numbers = {1, 2, 3, 2, 1}        # set

print(f"Fruits: {fruits}")
print(f"Coordinates: {coordinates}")
print(f"Student Grades: {student_grade}")
print(f"Unique Numbers: {unique_numbers}")
```
