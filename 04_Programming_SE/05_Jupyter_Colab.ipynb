## Core Concepts
Jupyter Notebook

This is an interactive environment that runs on your local machine. It allows you to create and share documents (called "notebooks") that contain live code, equations, visualizations, and narrative text. It's extremely popular in data science because it lets you run small pieces of code, see the results immediately, and document your thought process in one place.



Google Colab

Google Colab is essentially a Jupyter Notebook that runs in the cloud. The key advantages are:

- Zero setup: You can start coding right from your browser without installing anything.
- Free resources: It provides free access to powerful hardware like GPUs (Graphics Processing Units) and TPUs (Tensor Processing Units), which can dramatically speed up machine learning and data analysis tasks.
- Easy collaboration: Notebooks are saved to your Google Drive, making them as easy to share and edit as a Google Doc.


Cells

Notebooks are composed of individual blocks called cells. There are two primary types:

- Code Cells: These contain code (e.g., Python) that you can execute. The output of the code (text, plots, tables) is displayed directly below the cell.
- Markdown Cells: These contain text formatted using Markdown. You use these cells to write descriptions, add headings, embed images, and structure your notebook to make it readable and easy to understand.

Markdown

Markdown is a lightweight markup language for creating formatted text using a plain-text editor. It's how you write rich text in notebook cells. For example:

- Heading 1 creates a large heading.
- *italic text* or _italic text_ makes text italic.
- **bold text** or __bold text__ makes text bold.
- A list item creates a bulleted list.

Magic Commands

These are special commands in Jupyter and Colab that are not part of the Python language itself but provide useful shortcuts. They are prefixed with a % (for a single line) or %% (for a whole cell).

- %lsmagic: Lists all available magic commands.
- %timeit: Measures how long it takes for a piece of code to run.
- %%bash: Runs the entire cell's content as a shell script.
- %matplotlib inline: Ensures that plots generated by libraries like Matplotlib are displayed directly within the notebook.
