## Core Concepts

Variables and Data Types

Variables are names that store data values. Data types are categories for these values, like numbers (integers, floats), text (strings), and true/false booleans.
Control Flow

This directs the order of execution. if statements run code only if a condition is true. Loops (for, while) repeat code multiple times.

Functions

Reusable blocks of code that perform a specific task. You define a function once and can then call (execute) it whenever needed.

Lists and Dictionaries


    - Lists: Ordered collections of items, accessed by their numerical index (e.g., my_list[0]).
    - Dictionaries: Unordered collections of key-value pairs, accessed by a unique key (e.g., my_dict['name']).

Importing Modules

This allows you to use code written in other files (modules), giving you access to pre-built functions and variables.

## Example

```python
# Simple Python script
def greet(name):
    return f"Hello, {name}! Welcome to AI."

names = ['Alice', 'Bob', 'Charlie']
for name in names:
    print(greet(name))
```
