## Core Concepts
Functions

Functions are named, reusable blocks of code that perform a specific action. You define a function once and can then "call" it whenever you need to execute that code, which helps avoid repetition.

```python
def greet(name):
  """This function greets the person passed in as a parameter."""
  print(f"Hello, {name}!")

# Calling the function
greet("Alice")  # Output: Hello, Alice!
```

Modules

A module is simply a Python file with a .py extension. Modules allow you to logically organize your code by grouping related functions, classes, and variables into separate files.

Example: If you save the greet function above in a file named greetings.py, you've created a module.

Imports

The import statement is used to bring the code from one module into another. This allows you to use functions and variables defined in different files.

Example: To use the greet function from greetings.py in another file, you would do this:

```python
import greetings

greetings.greet("Bob") # Output: Hello, Bob!
```

Docstrings and Comments

These are ways to document your code, but they serve different purposes.

Docstrings ("""..."""): Explain what a function, module, or class does. They are placed as the first statement and can be accessed programmatically.
Comments (#): Explain the why or how of a specific line or block of code. They are for developers reading the code and are ignored by the Python interpreter.

```python
# Example function with docstring
def add_numbers(a, b):
    """Return the sum of two numbers."""
    return a + b

result = add_numbers(5, 7)
print(f"Sum: {result}")
```

Testing

Testing is the process of writing code to verify that your other code works correctly. Simple tests can be written using the assert statement, which checks if a condition is true and raises an error if it's not.

Example: To test the calculate_area function:

```python
# A simple test
assert calculate_area(10) == 314.159

# If the function works correctly, this line will run without any output.
# If the result is not 314.159, it will raise an AssertionError.
```
