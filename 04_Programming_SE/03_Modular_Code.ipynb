## Core Concepts
- **Functions:** Defining reusable code blocks.
- **Modules:** Organizing functions and classes into files.
- **Imports:** Using code across different files.
- **Docstrings and Comments:** Documenting code effectively.
- **Testing:** Writing simple tests for code correctness.

## Example

```python
# Example function with docstring
def add_numbers(a, b):
    """Return the sum of two numbers."""
    return a + b

result = add_numbers(5, 7)
print(f"Sum: {result}")
```
