## Introduction
Visualization helps convert raw data into insight by making patterns and anomalies visible. This notebook covers essential visualization techniques for AI data analysis.

## Core Concepts
- **Histograms:** Visualize distribution of numerical data.
- **Scatter Plots:** Explore relationships between variables.
- **Box Plots:** Show distribution summary and identify outliers.
- **Bar Charts:** Compare categories.
- **Pair Plots:** Visualize pairwise relationships in datasets.

## Example
Plotting with Matplotlib and Seaborn.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Load example dataset
df = sns.load_dataset("iris")

# Histogram of sepal length
plt.figure(figsize=(8, 4))
plt.hist(df['sepal_length'], bins=20, color='skyblue')
plt.title('Histogram of Sepal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')
plt.show()

# Pair plot
sns.pairplot(df, hue='species')
```
