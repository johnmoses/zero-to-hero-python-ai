## Introduction
Data collection is the process of gathering data, information or any variables of interest for the purpose of hypothesis testing and evaluation of outcomes. Pandas can read data from many sources.

The quality and quantity of data significantly affect AI system performance. This notebook covers different methodologies to acquire data for AI projects.

## Core Concepts
- **Manual Data Entry:** Human-generated data inputs.
- **Automated Data Capture:** Sensors, logs, web scraping.
- **APIs and Public Datasets:** Using external data sources.
- **Surveys and Experiments:** Structured data collection.
- **Data Privacy & Ethics:** Legal and ethical considerations.

## Example
- Fetching data using an API


```python
# Example: Simple API request using requests module (internet required)
import requests

url = 'https://api.agify.io?name=michael'
response = requests.get(url)
if response.status_code == 200:
    data = response.json()
    print(f"Name: {data['name']}, Predicted Age: {data['age']}")
else:
    print('Failed to retrieve data')
```
