## Introduction
Data collection is the process of gathering data, information or any variables of interest for the purpose of hypothesis testing and evaluation of outcomes. Pandas can read data from many sources.

The quality and quantity of data significantly affect AI system performance. This notebook covers different methodologies to acquire data for AI projects.

## Core Concepts

- Manual Data Entry: This involves a person manually typing or recording data into a system, such as filling out a form or transcribing information. While it allows for collecting highly specific or nuanced information, it is often slow, expensive, and prone to human error (e.g., typos, inconsistencies).

- Automated Data Capture: This is the process of collecting data automatically without human intervention.
    
    - Sensors: Devices that measure physical properties like temperature, location (GPS), or motion.
    - Logs: System-generated records of events, such as server access logs, application errors, or user interactions on a website.
    - Web Scraping: Automatically extracting large amounts of data from websites. This method is fast, scalable, and often provides real-time data.

- APIs and Public Datasets:
    
    - APIs (Application Programming Interfaces): Allow different software systems to communicate and exchange data in a structured way. Companies often provide APIs for accessing their data (e.g., weather data, stock prices, social media trends).
    - Public Datasets: Curated collections of data released by governments, academic institutions, and other organizations for public use (e.g., census data, economic indicators). These are valuable for research and building models without needing to collect data from scratch.

- Surveys and Experiments: These are proactive methods for creating new data.
    - Surveys: Collect information from a sample of individuals through a series of questions, useful for gathering opinions, preferences, and demographic data.
    - Experiments: Involve manipulating variables in a controlled environment to observe the effect on an outcome (e.g., A/B testing different website layouts to see which one leads to more clicks).

- Data Privacy & Ethics: This involves handling data responsibly.
    - Legal: Complying with regulations like GDPR or CCPA, which govern how personal data can be collected, stored, and used.
    - Ethical: Going beyond legal requirements to ensure fairness and respect for individuals. This includes obtaining informed consent, anonymizing data to protect identities, being transparent about how data is used, and actively working to prevent biases in data collection and model training that could lead to unfair outcomes.

## Example
- Fetching data using an API

```python
# Example: Simple API request using requests module (internet required)
import requests

url = 'https://api.agify.io?name=michael'
response = requests.get(url)
if response.status_code == 200:
    data = response.json()
    print(f"Name: {data['name']}, Predicted Age: {data['age']}")
else:
    print('Failed to retrieve data')
```
