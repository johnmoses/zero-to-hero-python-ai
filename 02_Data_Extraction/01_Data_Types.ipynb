## Introduction
Data is the foundation of AI. Different types of data require different processing and modeling techniques. This notebook introduces the major data types you will encounter in AI projects.

## Core Concepts
- **Structured Data:** Organized data that fits neatly into rows and columns, like databases and spreadsheets (e.g., CSV, SQL tables).
- **Unstructured Data:** Data without a predefined structure, such as images, audio, videos, and free-text documents.
- **Semi-Structured Data:** Data that does not fit rigid tables but has some organizational properties, such as JSON, XML, and HTML.
- **Time Series Data:** Data points indexed in time order.
- **Categorical vs Numerical Data:** Important distinctions for modeling.

## Examples
- Customer databases, e-commerce transactions (structured)
- Social media posts, videos, medical images (unstructured)
- Log files, emails (semi-structured)

```python
# Sample Python code demonstrating structured vs unstructured data
import pandas as pd

# Structured data example
data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Age': [25, 30, 35]}
df = pd.DataFrame(data)
print("Structured Data (DataFrame):")
print(df)

# Unstructured data example (text)
text = "Artificial intelligence is transforming industries."
print("\nUnstructured Data (Text):")
print(text)
```
