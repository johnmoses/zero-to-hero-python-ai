# Data Cleaning and Handling Missing Values

## Learning Objectives
- Understand the importance of data cleaning in AI workflows.
- Learn common techniques to handle missing or inconsistent data.
- Practice cleaning datasets with Python and pandas.

## Introduction
Real-world data is often messy: it may have missing values, inconsistencies, or errors. Cleaning data is a critical step that directly affects model quality and performance.

## Core Concepts
- **Identifying Missing Data:** Using methods to detect missing or null values.
- **Handling Missing Data:** Strategies include removal, imputation (mean, median, mode), and interpolation.
- **Dealing with Outliers:** Identifying and addressing anomalies.
- **Data Consistency:** Standardizing formats and correcting errors.
- **Data Validation:** Ensuring dataset meets quality constraints.

## Example
Use pandas to detect and fill missing data.

```python
import pandas as pd
import numpy as np

# Sample dataset with missing values
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, np.nan, 30, 22],
        'Salary': [50000, 60000, np.nan, 40000]}
df = pd.DataFrame(data)
print("Original Data:")
print(df)

# Detect missing values
print("\nMissing Values per Column:")
print(df.isnull().sum())

# Fill missing Age with mean
df['Age'].fillna(df['Age'].mean(), inplace=True)

# Fill missing Salary with median
df['Salary'].fillna(df['Salary'].median(), inplace=True)

print("\nData after Imputation:")
print(df)
```
