# Data Cleaning and Handling Missing Values

## Learning Objectives

## Introduction
Real-world data is often messy: it may have missing values, inconsistencies, or errors. Cleaning data is a critical step that directly affects model quality and performance.

## Core Concepts

- Identifying Missing Data: This is the first step in cleaning a dataset. It involves programmatically scanning the data (e.g., using functions like isnull() or isna() in libraries like Pandas) to find empty or null entries (often represented as NaN, None, or NULL). Missing data can skew analysis and cause errors in machine learning models.

- Handling Missing Data: Once identified, you must decide how to handle the gaps.
    - Removal: The simplest strategy is to delete rows with missing values (listwise deletion) or entire columns if they are mostly empty. This is only suitable if the amount of missing data is small, as it can lead to significant information loss.
    - Imputation: This involves filling in the missing values. Common methods include replacing them with the mean, median (more robust to outliers), or mode (for categorical data) of the respective column.
    - Interpolation: For ordered data like time series, you can estimate the missing value based on the values of its neighbors.

- Dealing with Outliers: Outliers are data points that are significantly different from other observations. They can be identified using statistical methods (e.g., values outside 3 standard deviations from the mean) or visualization (e.g., box plots). Outliers can be legitimate data points or errors. Depending on the context, they can be removed, transformed (e.g., using a log scale), or imputed.

- Data Consistency: This involves ensuring that data is uniform and free of contradictions. It includes standardizing formats (e.g., converting all dates to YYYY-MM-DD), correcting typos, and ensuring categorical values are consistent (e.g., mapping "USA", "U.S.A.", and "United States" to a single, standard entry).

- Data Validation: This is the process of applying rules to ensure the overall quality and integrity of the data. It involves checking that values fall within expected ranges (e.g., age should be positive), that data types are correct (e.g., a 'price' column contains numbers, not text), and that the data adheres to predefined business logic before it is used for analysis or modeling.

## Example
Use pandas to detect and fill missing data.

```python
import pandas as pd
import numpy as np

# Sample dataset with missing values
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, np.nan, 30, 22],
        'Salary': [50000, 60000, np.nan, 40000]}
df = pd.DataFrame(data)
print("Original Data:")
print(df)

# Detect missing values
print("\nMissing Values per Column:")
print(df.isnull().sum())

# Fill missing Age with mean
df['Age'].fillna(df['Age'].mean(), inplace=True)

# Fill missing Salary with median
df['Salary'].fillna(df['Salary'].median(), inplace=True)

print("\nData after Imputation:")
print(df)
```
