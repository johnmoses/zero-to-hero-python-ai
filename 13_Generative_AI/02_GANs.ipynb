## Introduction
GANs train two networks adversarially: the generator tries to produce realistic data, while the discriminator tries to distinguish real from fake.

## Core Concepts
- **Generator:** Produces synthetic data.
- **Discriminator:** Classifies real vs fake data.
- **Adversarial Loss:** Drives the competing objectives.
- **Training Challenges:** Mode collapse, instability.
