## Introduction
Generative AI powers tasks such as creating realistic images, generating human-like text, and aiding scientific discoveries.

## Core Applications
- **Image Generation:** Creating photorealistic images (e.g., GANs).
- **Text Generation:** Chatbots and creative writing.
- **Data Augmentation:** Enhancing datasets with synthetic samples.
- **Drug Discovery:** Generating new molecular structures.

## Example
Image generation using pretrained GANs is demonstrated.

```python
# This example assumes access to a pretrained GAN generator model
# Pseudocode only (no actual generator provided)
import numpy as np
import matplotlib.pyplot as plt

# Generate random latent vectors
latent_dim = 100
random_latent_vectors = np.random.normal(size=(5, latent_dim))

# Generate images (placeholder function)
def generate_images(latent_vectors):
    # Pretend to generate images from GAN
    images = np.random.rand(latent_vectors.shape[0], 64, 64, 3)
    return images

images = generate_images(random_latent_vectors)

# Display images
for i, img in enumerate(images):
    plt.subplot(1, 5, i+1)
    plt.imshow(img)
    plt.axis('off')
plt.show()
```
