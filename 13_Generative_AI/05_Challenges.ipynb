# Challenges in Generative AI

## Learning Objectives
- Discover major challenges faced by generative AI models.
- Understand ethical, technical, and practical issues.

## Introduction
Generative AI presents opportunities but also challenges including bias, mode collapse, and misuse.

## Core Challenges
- **Mode Collapse:** Generator produces limited variety.
- **Training Instability:** Difficult and sensitive optimization.
- **Bias and Fairness:** Generated outputs may amplify biases.
- **Ethical Concerns:** Deepfakes and misinformation.
- **Computational Costs:** High resource requirements.
