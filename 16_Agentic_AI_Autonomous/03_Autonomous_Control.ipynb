## Introduction
Autonomous control enables agents to perform tasks without human intervention using control theory and AI.

## Core Concepts

- Feedback Control: A method where the output of a system is measured and compared to a desired setpoint, and the difference (error) is used to adjust the system's input.

    
    - Mechanism: It creates a "closed loop" that continuously works to minimize the error between the current state and the desired state.
    
    - Goal: To automatically regulate a system and make it robust against unexpected disturbances.
    
    - Key Components: A sensor (to measure), a controller (to decide the action), and an actuator (to perform the action).
    
    - Classic Example: A cruise control system in a car. It measures your actual speed, compares it to the set speed, and adjusts the throttle to correct any error, especially when going up or down hills.

- Model Predictive Control (MPC): An advanced control strategy that uses a model of the system to predict its future behavior and calculate the optimal sequence of control actions.

    
    - Mechanism: At every step, it solves an optimization problem for a future time window (the "prediction horizon"). It calculates the best plan but only implements the first step of that plan.
    
    - Receding Horizon: After implementing the first step, it re-measures the system's state and repeats the whole process, creating a new optimal plan from the new state.
    
    - Key Trait: Excellent for handling complex systems with multiple variables and constraints (e.g., safety limits, actuator limits).
    
    - Analogy: Playing chess. You think several moves ahead to find the best strategy, but you only make one move. Then you see your opponent's move (the new system state) and recalculate your strategy from there.

- Adaptive Control: A type of control that can modify its own parameters in real-time to handle systems whose properties change or are uncertain.

    
    - Mechanism: It runs a standard feedback control loop while a second "adaptation loop" continuously estimates the system's current dynamics and updates the controller to match.
    
    - Goal: To maintain performance and stability even when the system behaves in unexpected ways or its characteristics drift over time.
    
    - When to Use: Ideal for systems that are poorly understood or whose dynamics change during operation (e.g., an aircraft's handling characteristics changing as it burns fuel).
    
    - Analogy: A person adjusting their posture and balance when switching from carrying an empty backpack to a very heavy one. Your internal control system adapts to the new weight and its distribution.

## Example
Simple proportional controller pseudocode.

```python
def proportional_controller(setpoint, measured_value, Kp=1.0):
    error = setpoint - measured_value
    control_signal = Kp * error
    return control_signal

print(proportional_controller(10, 7))
```
