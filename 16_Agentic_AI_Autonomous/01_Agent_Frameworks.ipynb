## Introduction
Agent frameworks enable AI systems to perceive, reason, and act autonomously in complex environments.

## Core Architectures

- Reactive Agents: Respond directly to the environment based on pre-programmed rules.

    - Mechanism: Uses simple IF-THEN logic (e.g., IF obstacle detected, THEN turn).
    - Key Trait: Extremely fast and simple, acting like a reflex.
    - Limitation: Has no memory or ability to plan ahead.
    - Analogy: A thermostat reacting to the current temperature.

- Deliberative Agents: Use an internal model of the world to think ahead and create a step-by-step plan.

    - Mechanism: Follows a "Sense-Plan-Act" cycle to achieve a specific goal.
    - Key Trait: Can solve complex problems that require foresight.
    - Limitation: Can be very slow to react, as planning takes time.
    - Analogy: A GPS app calculating the best route before a trip.

- Hybrid Agents: Combine reactive and deliberative approaches to get the best of both.

    - Mechanism: Uses a fast, reactive layer for emergencies and a deliberative layer for long-term goals.
    - Key Trait: Balances being goal-oriented with being responsive to the environment.
    - Benefit: More robust and versatile than a pure reactive or deliberative agent.
    - Analogy: A driver following a planned route who also reacts instantly to avoid an accident.

- BDI Model: A framework for building a "thinking" agent that mimics human practical reasoning.

    - Mechanism: The agent makes decisions by reasoning about its mental state, which is defined by:
        - Beliefs: What it knows about the world.
        - Desires: What it wants to achieve (all possible goals).
        - Intentions: The goals it has committed to and is actively pursuing.
    - Analogy: The process of deciding what to cook based on what's in your fridge (beliefs) and what you're in the mood for (desires).

## Example
Simple rule-based reactive agent pseudocode.

```python
def reactive_agent(percept):
    if percept == 'obstacle':
        action = 'turn'
    else:
        action = 'move_forward'
    return action

print(reactive_agent('clear'))
print(reactive_agent('obstacle'))
```
