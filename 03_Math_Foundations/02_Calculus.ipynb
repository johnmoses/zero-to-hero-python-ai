# Calculus for Machine Learning

## Learning Objectives
- Understand basic calculus concepts relevant to AI.
- Learn about derivatives, gradients, and their role in optimization.

## Introduction
Calculus, especially differential calculus, is fundamental in training machine learning models. It helps in understanding how models learn by optimizing loss functions using gradients.

## Functions
A Function is a mapping between and input set and an output set. An element of the output set is related to exactly one element of the output set for example f(x) = x2

## Core Concepts
- **Derivatives:** Rate of change of a function.
- **Partial Derivatives:** Derivative of multivariate functions with respect to one variable.
- **Gradient:** Vector of partial derivatives indicating steepest ascent direction.
- **Gradient Descent:** Algorithm for minimizing functions using gradients.
- **Chain Rule:** Computing derivatives of composite functions.

## Example
Compute derivative and gradient using SymPy.

```python
import sympy as sp

# Define variables
x, y = sp.symbols('x y')

# Define function
f = x**2 + y**2

# Compute partial derivatives
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)
gradient = [df_dx, df_dy]

print(f"Function: {f}")
print(f"Derivative w.r.t x: {df_dx}")   # With respect to x
print(f"Derivative w.r.t y: {df_dy}")   # With respect to y
print(f"Gradient: {gradient}")
```

Partial Derivatives

```python
from sympy import symbols, diff, init_printing, sin, exp

# Use init_printing for beautifully rendered math output in Jupyter
init_printing(use_unicode=True)

# 1. Define the symbolic variables you will use
x, y, z = symbols('x y z')

# 2. Define a multivariate function
# Let's use f(x, y) = x^4 * y + sin(x*y)
f = x**4 * y + sin(x*y)

# To see the function beautifully rendered, just put it as the last line in a cell
print("Original function f(x, y):")
f
```

Gradient

```python
# Compute the gradient (partial derivatives with respect to x and y)
grad_x = diff(f, x)
grad_y = diff(f, y)
gradient = [grad_x, grad_y]

print("Gradient of f(x,y):")
print(f"∂f/∂x = {grad_x}")
print(f"∂f/∂y = {grad_y}")
```

Gradient Descent

```python
import numpy as np

# Implement gradient descent
def gradient_descent(start_x, start_y, learning_rate=0.01, n_iterations=100):
    # Current position
    x_current = float(start_x)
    y_current = float(start_y)
    
    # Store history for visualization
    history = [(x_current, y_current)]
    
    for i in range(n_iterations):
        # Evaluate gradients at current point
        dx = float(grad_x.subs([(x, x_current), (y, y_current)]))
        dy = float(grad_y.subs([(x, x_current), (y, y_current)]))
        
        # Update position
        x_current = x_current - learning_rate * dx
        y_current = y_current - learning_rate * dy
        
        # Store new position
        history.append((x_current, y_current))
    
    return x_current, y_current, history

# Run gradient descent starting from point (2, 2)
final_x, final_y, path = gradient_descent(2, 2)
print(f"Minimum found at: x = {final_x:.4f}, y = {final_y:.4f}")
```

Chain Rule

```python
# Create composite functions to demonstrate chain rule
# Let's use h(x,y) = f(g(x,y)) where g(x,y) = x^2 + y^2

# Define g(x,y)
g = x**2 + y**2

# Create a composite function h(x,y) = f(g(x,y))
# Using our existing f(x,y) = x^4*y + sin(x*y)
h = f.subs([(x, g), (y, y)])

# Calculate derivatives using chain rule
dh_dx = diff(h, x)
dh_dy = diff(h, y)

print("Composite function h(x,y):")
print(h)
print("\nPartial derivative with respect to x (using chain rule):")
print(dh_dx)
print("\nPartial derivative with respect to y (using chain rule):")
print(dh_dy)
```
