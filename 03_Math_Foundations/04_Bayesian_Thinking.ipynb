# Bayesian Thinking in AI

## Learning Objectives
- Understand Bayes' theorem and its role in AI.
- Learn how to update beliefs using evidence.

## Introduction
Bayesian thinking provides a framework for reasoning under uncertainty and is a key concept in AI models such as probabilistic graphical models and Bayesian networks.

## Core Concepts
- **Bayes' Theorem:** Formula to update probability estimates based on new evidence.
- **Prior, Likelihood, Posterior:** Components of Bayesian inference.
- **Conditional Probability:** Probability of an event given that another event has occurred.

## Example
Calculate posterior probabilities given priors and likelihoods.

```python
# Example: Medical test example
prior_disease = 0.01  # 1% disease prevalence
likelihood_positive_if_disease = 0.99
likelihood_positive_if_no_disease = 0.05

# Apply Bayes' Theorem
numerator = likelihood_positive_if_disease * prior_disease
denominator = numerator + likelihood_positive_if_no_disease * (1 - prior_disease)
posterior = numerator / denominator

print(f"Probability of disease given positive test: {posterior:.4f}")
```

Conditional probabiity

```python
import numpy as np

import matplotlib.pyplot as plt

# Function to calculate conditional probability using Bayes' theorem
def conditional_probability(prior, likelihood_positive, likelihood_negative):
    numerator = likelihood_positive * prior
    denominator = numerator + likelihood_negative * (1 - prior)
    return numerator / denominator

# Create range of prior probabilities to visualize
prior_range = np.linspace(0.01, 0.5, 100)
posteriors = [conditional_probability(p, likelihood_positive_if_disease, 
                                   likelihood_positive_if_no_disease) 
              for p in prior_range]

# Plot the relationship between prior and posterior probabilities
plt.figure(figsize=(10, 6))
plt.plot(prior_range, posteriors, 'b-', label='Posterior Probability')
plt.plot(prior_range, prior_range, 'r--', label='Prior=Posterior Line')
plt.xlabel('Prior Probability of Disease')
plt.ylabel('Posterior Probability of Disease')
plt.title('Conditional Probability: Prior vs Posterior')
plt.legend()
plt.grid(True)
```

Baye's theorem

```python
def bayes_theorem(prior, sensitivity, specificity):
    """
    Calculate posterior probability using Bayes' theorem
    
    Args:
        prior: Prior probability of event
        sensitivity: True Positive Rate (P(positive|condition))
        specificity: True Negative Rate (P(negative|no condition))
    """
    likelihood_positive = sensitivity
    likelihood_negative = 1 - specificity
    
    # Calculate posterior using Bayes' theorem
    numerator = likelihood_positive * prior
    denominator = (likelihood_positive * prior) + (likelihood_negative * (1 - prior))
    posterior = numerator / denominator
    
    return posterior

# Test cases with different prior probabilities
test_cases = [
    {'prior': 0.01, 'name': 'Rare condition'},
    {'prior': 0.3, 'name': 'Common condition'},
    {'prior': 0.5, 'name': 'Equal probability'}
]

# Using the existing sensitivity and specificity from previous cells
sensitivity = likelihood_positive_if_disease  # 0.99
specificity = 1 - likelihood_positive_if_no_disease  # 0.95

# Calculate and display results
for case in test_cases:
    post_prob = bayes_theorem(case['prior'], sensitivity, specificity)
    print(f"\n{case['name']}:")
    print(f"Prior probability: {case['prior']:.2f}")
    print(f"Posterior probability: {post_prob:.4f}")

# Visualize how posterior changes with different priors
new_prior_range = np.linspace(0.01, 0.99, 100)
new_posteriors = [bayes_theorem(p, sensitivity, specificity) for p in new_prior_range]

plt.figure(figsize=(10, 6))
plt.plot(new_prior_range, new_posteriors, 'b-', label='Posterior Probability')
plt.plot([0, 1], [0, 1], 'r--', label='Prior=Posterior Line')
plt.xlabel('Prior Probability')
plt.ylabel('Posterior Probability')
plt.title("Bayes' Theorem: Prior vs Posterior Probability")
plt.legend()
plt.grid(True)
plt.tight_layout()
```
