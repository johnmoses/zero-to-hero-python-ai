## Introduction
Bayesian thinking provides a framework for reasoning under uncertainty and is a key concept in AI models such as probabilistic graphical models and Bayesian networks.

## Core Concepts
- **Bayes' Theorem:** Formula to update probability estimates based on new evidence.
- **Prior, Likelihood, Posterior:** Components of Bayesian inference.
- **Conditional Probability:** Probability of an event given that another event has occurred.

## Example
Calculate posterior probabilities given priors and likelihoods.

```python
# Example: Medical test example
prior_disease = 0.01  # 1% disease prevalence
likelihood_positive_if_disease = 0.99
likelihood_positive_if_no_disease = 0.05

# Apply Bayes' Theorem
numerator = likelihood_positive_if_disease * prior_disease
denominator = numerator + likelihood_positive_if_no_disease * (1 - prior_disease)
posterior = numerator / denominator

print(f"Probability of disease given positive test: {posterior:.4f}")
```
