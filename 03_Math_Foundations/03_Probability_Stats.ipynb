## Introduction
Probability and statistics provide the mathematical framework to model uncertainty and make inferences from data, which is essential in AI systems.

## Core Concepts
- **Probability:** Likelihood of events occurring.
- **Random Variables:** Variables whose outcomes are uncertain.
- **Distributions:** Ways probabilities are distributed over outcomes (e.g., Normal, Bernoulli).
- **Expectation and Variance:** Measures of central tendency and spread.
- **Statistical Inference:** Deduce properties of populations from samples.

## Example
Calculate mean, variance, and create a simple probability distribution.

```python
import numpy as np

# Sample data
data = np.array([2, 4, 4, 4, 5, 5, 7, 9])

# Calculate mean and variance
mean = np.mean(data)
variance = np.var(data)
print(f"Mean: {mean}")
print(f"Variance: {variance}")

# Create a Normal distribution sample
samples = np.random.normal(loc=mean, scale=np.sqrt(variance), size=1000)
print(f"Generated {len(samples)} samples from a normal distribution.")
```
