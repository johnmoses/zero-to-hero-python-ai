## Introduction
AI combines multiple branches of mathematics — linear algebra, calculus, probability, and optimization — to build models that can learn from data and make decisions.

## Core Connections
- **Linear Algebra:** Data representation and model parameters.
- **Calculus:** Optimization and training of models.
- **Probability/Statistics:** Handling uncertainty and inference.
- **Optimization:** Minimizing loss functions to find best model parameters.

## Example: Linear Regression
Combines linear algebra (matrix operations), calculus (gradient computation), and statistics (estimating parameters) to fit a model.

```python
import numpy as np

# Simple linear regression using normal equation
X = np.array([[1, 1], [1, 2], [2, 2], [2, 3]])  # features with intercept
y = np.dot(X, np.array([1, 2])) + 3  # y = 1* x1 + 2*x2 + 3

# Calculate parameters (theta) with normal equation
theta = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
print(f"Estimated parameters: {theta}")
```

Optimization

```python
# Implement gradient descent optimization
learning_rate = 0.01
num_iterations = 1000
m = len(y)

# Initialize parameters
theta_gd = np.zeros(2)

# Gradient descent
for i in range(num_iterations):
    # Compute predictions
    y_pred = np.dot(X, theta_gd)
    
    # Compute gradients
    gradients = (1/m) * X.T.dot(y_pred - y)
    
    # Update parameters
    theta_gd = theta_gd - learning_rate * gradients

print(f"Gradient descent parameters: {theta_gd}")
print(f"Mean squared error: {np.mean((np.dot(X, theta_gd) - y)**2)}")

# Compare with the analytical solution we already have
print(f"Analytical solution parameters: {theta}")
print(f"Mean squared error (analytical): {np.mean((np.dot(X, theta) - y)**2)}")
```
