# Gradient Descent

## Learning Objectives
- Understand the gradient descent optimization algorithm.
- Learn the update rules and convergence criteria.

## Introduction
Gradient descent is an iterative optimization algorithm used to minimize a loss function by updating parameters in the direction of the negative gradient.

## Core Concepts
- **Learning Rate:** Step size for updates.
- **Batch Gradient Descent:** Uses entire dataset per update.
- **Stochastic Gradient Descent (SGD):** Uses single samples.
- **Mini-batch Gradient Descent:** Uses small batches.

## Example
Update weights using gradient descent.

```python
import numpy as np

def gradient_descent_step(w, grad_w, learning_rate):
    return w - learning_rate * grad_w

# Example
w = np.array([1.0, 2.0])
grad_w = np.array([0.1, -0.2])
learning_rate = 0.05

w_new = gradient_descent_step(w, grad_w, learning_rate)
print(f"Updated weights: {w_new}")
```
