# Batch Normalization and Layer Normalization

## Learning Objectives
- Understand batch normalization and layer normalization techniques.
- Learn how normalization improves training stability and speed.

## Introduction
Normalization methods reduce internal covariate shift and smooth optimization landscapes.

## Core Concepts
- **Batch Normalization:** Normalizes inputs over the batch dimension.
- **Layer Normalization:** Normalizes inputs over features per example.
- **Effects:** Faster convergence, reduced sensitivity to initialization.

## Example
Apply Batch Normalization in a TensorFlow model.

```python
import tensorflow as tf
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Dense(64, activation='relu'),
    layers.BatchNormalization(),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()
```
