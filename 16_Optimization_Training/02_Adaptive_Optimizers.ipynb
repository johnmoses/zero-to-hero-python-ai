# Adaptive Optimizers

## Learning Objectives
- Understand adaptive optimization algorithms like Adam, RMSProp, and Adagrad.
- Learn advantages and applications.

## Introduction
Adaptive optimizers adjust learning rates for each parameter, improving training speed and convergence in many scenarios.

## Core Concepts
- **Adagrad:** Adjusts learning rates based on past gradients.
- **RMSProp:** Uses moving average of squared gradients.
- **Adam:** Combines momentum and RMSProp.

## Example
Using Adam optimizer in TensorFlow.

```python
import tensorflow as tf

optimizer = tf.keras.optimizers.Adam(learning_rate=0.001)

# Example training step (pseudocode)
def train_step(model, inputs, labels):
    with tf.GradientTape() as tape:
        predictions = model(inputs)
        loss = tf.keras.losses.sparse_categorical_crossentropy(labels, predictions)
    grads = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    return loss
```
