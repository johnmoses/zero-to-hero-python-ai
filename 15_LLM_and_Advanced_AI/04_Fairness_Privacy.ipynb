# Fairness and Privacy in LLMs

## Learning Objectives
- Understand fairness concerns in large language models.
- Learn methods to improve privacy and reduce bias.

## Introduction
Increasing deployment of LLMs raises ethical issues including bias, fairness, and user privacy.

## Core Issues

- Bias: Since models are trained on human-generated data from the internet, they inevitably learn and internalize the social, historical, and cultural biases present in that data. This can manifest as stereotyping, prejudiced associations, or generating text that is demeaning to certain groups.

- Fairness: This is the goal of ensuring that a model's behavior and outcomes are equitable across different user groups (e.g., defined by race, gender, or other demographics). A model that exhibits bias is, by definition, unfair. For example, a loan application model that gives systematically lower approval scores to one demographic group than another, despite similar qualifications, would be considered unfair.

- Privacy: Training datasets are often scraped from the web and can contain sensitive, personal information (like names, addresses, or private conversations) or proprietary data. There's a risk that a model might "memorize" this data and inadvertently reproduce it in its output, leading to significant privacy violations.

The Mitigations

- Data Curation: This is a foundational step to address bias. It involves extensive filtering of the training dataset to remove toxic, hateful, and explicitly biased content. It also includes efforts to balance the data to ensure more equitable representation.

- Differential Privacy: This is a powerful technique to protect privacy. It adds a carefully controlled amount of statistical noise during the training process. This makes it mathematically provable that the model's output is not dependent on any single individual's data, thus protecting the privacy of people whose data was in the training set.

- Bias Auditing: This is the process of actively testing a model for bias and fairness issues after it has been trained. It involves using specialized datasets and prompts to measure performance disparities across different demographic groups and to identify stereotypical or harmful associations the model may have learned. The results of these audits are used to guide further safety improvements.

## Example
Bias detection using embeddings.

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Example word embeddings (random vectors for illustration)
embedding_1 = np.array([0.1, 0.2, 0.3])
embedding_2 = np.array([0.2, 0.1, 0.4])
embedding_3 = np.array([0.9, 0.8, 0.7])

similarity_1_2 = cosine_similarity([embedding_1], [embedding_2])[0]
similarity_1_3 = cosine_similarity([embedding_1], [embedding_3])

print(f"Similarity between 1 and 2: {similarity_1_2}")
print(f"Similarity between 1 and 3: {similarity_1_3}")
```
