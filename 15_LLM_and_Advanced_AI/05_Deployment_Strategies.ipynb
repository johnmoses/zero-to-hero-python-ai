## Learning Objectives

## Introduction
Deploying LLMs requires engineering strategies to manage computational load, latency, and user privacy.

## Core Topics

- Model Serving: This is the process of making a trained model available for applications to use.

    - APIs/Microservices: The most common method for real-time applications. The model is wrapped in a dedicated service that exposes an API endpoint. Other applications can then send requests to this endpoint and get predictions back.
    
    - Batch Processing: Used for non-real-time tasks. The model runs on a schedule to process large volumes of data at once, saving the results to a database or file store.

- Scaling: This addresses how to handle large models and high request volumes.

    - Horizontal Scaling: To handle more users, you simply add more machines, each running a copy of the model, and use a load balancer to distribute the traffic.
    
    - Model Parallelism: For models that are too large to fit on a single GPU, the model itself is split across multiple GPUs. Processing a single request requires communication between these GPUs.

- Latency: This is the time it takes from sending a request to the model to receiving a response. For user-facing applications like chatbots or recommendation systems, low latency is critical for a good user experience. Managing it involves optimizing the model, using powerful hardware, and efficient request batching.

- Cost Optimization: Serving large models, especially on high-end GPUs, is very expensive. Cost optimization involves strategies to get the most performance for the lowest cost. This can include auto-scaling infrastructure to match demand, using more efficient model formats (quantization), and batching requests to maximize the utilization of the hardware.
