# Gradient Descent

## Learning Objectives

- Learn the update rules and convergence criteria.

## Introduction
Gradient descent is an iterative optimization algorithm used to minimize a loss function by updating parameters in the direction of the negative gradient.

## Core Concepts

- Learning Rate: A hyperparameter that controls the step size for updating the model's internal parameters (weights) during training.

    - Function: It determines how aggressively the model responds to the error it finds in each training step.
    - Too High: A large learning rate can cause the training to be unstable, overshooting the optimal solution.
    - Too Low: A small learning rate can make training extremely slow or cause the model to get stuck.
    - Analogy: Imagine walking down a hill in the fog to find the bottom. The learning rate is your step size. Big steps are fast but you might step right over the lowest point. Tiny steps are precise but take forever.

- Batch Gradient Descent: An optimization method that calculates the gradient using the entire training dataset to perform a single parameter update.

    - Advantage: Provides a very accurate and stable estimate of the gradient, leading to a smooth convergence.
    Disadvantage: Extremely slow and requires a lot of memory, making it impractical for the large datasets used in modern machine learning.
    - Analogy: A politician polling every single voter in the country before making one adjustment to their platform. The feedback is accurate, but the process is incredibly slow.

- Stochastic Gradient Descent (SGD): An optimization method that updates the parameters using the gradient calculated from just one single, randomly chosen training sample at a time.

Advantage: Very fast, allowing for rapid updates. The noisy steps can help the model escape poor local minima.
Disadvantage: The updates are erratic and have high variance, leading to a noisy, zig-zag path towards the solution.
Analogy: The politician asking one random person on the street for their opinion and immediately changing their platform. The feedback is fast but very unstable.

- Mini-batch Gradient Descent: The most common approach, which updates the parameters using a small, random batch of training samples (e.g., 32, 64, or 128 samples).

    - Function: It combines the benefits of the other two methods.
    - Advantage: More computationally efficient than Batch GD and provides a more stable gradient than SGD. It's the practical and effective compromise.
    - Analogy: The politician polling a small, random focus group. It's much faster than polling everyone but gives more stable feedback than asking just one person.

## Example
Update weights using gradient descent.

```python
import numpy as np

def gradient_descent_step(w, grad_w, learning_rate):
    return w - learning_rate * grad_w

# Example
w = np.array([1.0, 2.0])
grad_w = np.array([0.1, -0.2])
learning_rate = 0.05

w_new = gradient_descent_step(w, grad_w, learning_rate)
print(f"Updated weights: {w_new}")
```
