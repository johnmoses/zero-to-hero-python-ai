# Hyperparameter Tuning

## Learning Objectives
- Understand the importance of hyperparameters in model training.

## Introduction
Selecting optimal hyperparameters can significantly affect model performance and training efficiency.

## Core Concepts

- Hyperparameters

    - Definition: The configuration settings that are set before the training process begins. They are not learned from the data but are chosen by the practitioner to guide how the model learns.
    
    - Distinction: This is different from model parameters (like the weights in a neural network), which are learned automatically during training.
    - Examples:
        - Learning Rate: The step size for weight updates.
        - Batch Size: The number of samples used in one training iteration.
        - Number of Layers/Neurons: The architecture of the model.
        - Dropout Rate: The probability of dropping neurons for regularization.

- Grid Search

    - Method: An exhaustive search technique that systematically trains and evaluates a model for every possible combination of a specified grid of hyperparameter values.
    
    - Advantage: It is guaranteed to find the best combination within the provided grid.
    - Disadvantage: Extremely slow and computationally expensive. The number of combinations grows exponentially with the number of hyperparameters (the "curse of dimensionality").
    - Analogy: Searching for the best place to drill for oil by dividing a map into a perfect grid and drilling a well at every single intersection.

- Random Search

    - Method: A technique that randomly samples a fixed number of combinations from a specified range or distribution of hyperparameter values.
    
    - Advantage: Often finds better models than Grid Search in the same amount of time because it doesn't waste trials on unimportant parameters and explores the space more broadly.
    - Disadvantage: There is no guarantee that it will find the best combination.
    - Analogy: Searching for the best place to drill for oil by randomly dropping pins on the map and drilling where they land. You cover a wider, more diverse area than the grid search in the same number of attempts.

- Bayesian Optimization

    - Method: An "intelligent" search technique that uses the results from previous trials to decide which combination of hyperparameters to try next.
    - Mechanism: It builds a probabilistic model of the relationship between the hyperparameters and the model's performance. It then uses this model to select the next set of parameters that are most likely to yield an improvement.
    - Advantage: Far more efficient than Grid or Random Search. It finds better hyperparameters in significantly fewer iterations by making informed choices.
    - Analogy: Searching for the best place to drill for oil with an expert geologist. After each drill, the geologist analyzes the results and tells you the most promising place to drill next.

## Example
Using RandomizedSearchCV for hyperparameter tuning.

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
from scipy.stats import randint

model = RandomForestClassifier()
param_dist = {
    'n_estimators': randint(10, 200),
    'max_depth': randint(1, 20),
    'max_features': ['auto', 'sqrt', 'log2']
}
search = RandomizedSearchCV(model, param_distributions=param_dist, n_iter=10, cv=3)
# search.fit(X_train, y_train)  # Uncomment and provide data to run
print("RandomizedSearchCV instance created.")
```
