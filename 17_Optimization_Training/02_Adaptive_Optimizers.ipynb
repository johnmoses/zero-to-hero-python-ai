## Introduction
Adaptive optimizers adjust learning rates for each parameter, improving training speed and convergence in many scenarios.

## Core Concepts

- Adagrad (Adaptive Gradient Algorithm): An optimizer that adapts the learning rate for each parameter individually, giving larger updates to infrequent parameters and smaller updates to frequent ones.

    - Mechanism: It accumulates the squared gradients for each parameter over time and divides the learning rate by the square root of this sum.
    - Advantage: Works very well for sparse data (e.g., in natural language processing), where some features are seen much less often than others.
    - Disadvantage: The accumulated sum in the denominator only grows, causing the learning rate to eventually become infinitesimally small, effectively stopping the training process.

- RMSProp (Root Mean Square Propagation): An optimizer that fixes Adagrad's problem of a diminishing learning rate by using a moving average of squared gradients.

    - Mechanism: Instead of summing all past squared gradients, it uses an exponentially decaying average. This "forgets" the distant past and only considers recent gradient history.
    - Advantage: It prevents the learning rate from shrinking to zero and allows the model to continue learning effectively, making it a popular default choice.
    - Concept: It's like Adagrad but with a short-term memory, focusing only on the most recent gradient information.

- Adam (Adaptive Moment Estimation): The most common and often default optimizer, which combines the ideas of RMSProp with momentum.

    - Mechanism: It calculates two moving averages for each parameter:
    The average of the gradients (the momentum part, which helps accelerate in the correct direction).
    The average of the squared gradients (the RMSProp part, for adaptive per-parameter learning rates).
    - Advantage: It combines the best of both worlds—the fast convergence from momentum and the adaptive learning rates from RMSProp. It is generally robust and works well for a wide range of problems.
    - Concept: It's like a heavy ball rolling down a hill that not only builds up speed (momentum) but can also adjust its friction (learning rate) based on the steepness of the terrain.

## Example
Using Adam optimizer in TensorFlow.

```python
import tensorflow as tf

optimizer = tf.keras.optimizers.Adam(learning_rate=0.001)

# Example training step (pseudocode)
def train_step(model, inputs, labels):
    with tf.GradientTape() as tape:
        predictions = model(inputs)
        loss = tf.keras.losses.sparse_categorical_crossentropy(labels, predictions)
    grads = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    return loss
```
