# Learning Rate Schedules

## Learning Objectives
- Understand the importance of learning rate schedules.
- Learn different types of schedules: step decay, exponential decay, cosine annealing.

## Introduction
Adjusting the learning rate during training can improve convergence speed and model performance.

## Core Concepts

- Step Decay: A method that reduces the learning rate by a certain factor at pre-defined intervals (epochs).

    - Mechanism: The learning rate is held constant for a set number of epochs, then "steps down" to a lower value. This is repeated on a schedule.
    - Behavior: The learning rate plot looks like a staircase. This allows the model to make large, fast progress early on, then take smaller, more refined steps as it gets closer to a solution.
    - Analogy: Searching for a specific house. You drive fast to get to the right neighborhood, then drive slowly to find the right street, then crawl to find the exact house number.

- Exponential Decay: A method that smoothly decreases the learning rate after each epoch by a multiplicative factor.

    - Mechanism: The learning rate is multiplied by a decay factor (e.g., 0.95) after each epoch, resulting in a smooth, downward curve.
    - Behavior: The learning rate decreases more quickly at the beginning of training and then the rate of decay slows down over time.
    - Analogy: A cooling cup of coffee. It loses heat fastest when it's very hot, and the cooling process slows as it approaches room temperature.

- Cosine Annealing: A cyclical schedule that smoothly varies the learning rate from a high value to a low value following the curve of a cosine wave.

    - Mechanism: The learning rate starts high and is gradually "annealed" down to a minimum value over a set number of epochs. This cycle can then be restarted.
    - Behavior: The slow decrease allows the model to settle into a good minimum. The optional "warm restarts" (resetting the learning rate to a high value) can help the model pop out of a suboptimal solution and find a better one.
    - Analogy: A search team exploring a valley. They start by running down the hill (high learning rate), then slow to a walk to carefully explore the valley floor (low learning rate). They might then be airlifted to the top of a new hill to start the search again (a "warm restart").

## Example
Implement step decay learning rate schedule in TensorFlow.

```python
import tensorflow as tf

initial_learning_rate = 0.1
lr_schedule = tf.keras.optimizers.schedules.StepDecay(
    initial_learning_rate, decay_steps=10000, decay_rate=0.5, staircase=True)

optimizer = tf.keras.optimizers.SGD(learning_rate=lr_schedule)

print("Learning rate schedule created.")
```
