# Gradient Problems: Exploding and Vanishing Gradients

## Learning Objectives
- Understand the causes and effects of exploding and vanishing gradients.
- Learn techniques to mitigate these issues in training deep networks.

## Introduction
During backpropagation, gradients might become excessively small or large, harming effective learning.

## Core Concepts
- **Vanishing Gradients:** Gradients shrink, slowing or halting training of early layers.
- **Exploding Gradients:** Gradients grow uncontrollably, causing unstable updates.
- **Mitigation Techniques:** Gradient clipping, proper initialization, normalization, residual connections.

## Example
Gradient clipping in PyTorch.

```python
import torch
from torch import nn, optim

model = nn.Linear(10, 1)
optimizer = optim.SGD(model.parameters(), lr=0.1)

for input in [torch.randn(10) for _ in range(5)]:
    optimizer.zero_grad()
    output = model(input)
    loss = output.sum()
    loss.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
    optimizer.step()

print("Gradient clipping example executed.")
```
