# Batch Normalization and Layer Normalization

## Learning Objectives
- Understand batch normalization and layer normalization techniques.
- Learn how normalization improves training stability and speed.

## Introduction
Normalization methods reduce internal covariate shift and smooth optimization landscapes.

## Core Concepts

- Batch Normalization: A technique that normalizes the inputs to a layer by using the mean and standard deviation calculated from the current mini-batch.

    - Mechanism: For each feature, it calculates the mean and variance across all examples in the batch. It then standardizes the feature's values for that batch.

    - Use Case: Very common in Convolutional Neural Networks (CNNs) and other feed-forward networks.
    - Limitation: Its performance is dependent on the batch size; it works best when the batch is large enough to provide a good statistical estimate.
    - Analogy: A teacher grading on a curve for a specific class (the batch). The average and spread are calculated for that class, and all students in it are scaled accordingly.

- Layer Normalization: A technique that normalizes the inputs to a layer by using the mean and standard deviation calculated from all the features of a single training example.

    - Mechanism: For each individual example, it calculates the mean and variance across all of its features. It then uses these stats to standardize that single example's inputs.
    
    - Use Case: Very common in Recurrent Neural Networks (RNNs) and Transformers, where processing sequences of varying lengths makes batch statistics less reliable.
    - Advantage: It is independent of the batch size.
    - Analogy: A teacher normalizing a single student's report card. They calculate that one student's average grade across all their subjects and scale their grades based on their personal performance.

- Effects (of Normalization): These techniques significantly stabilize and accelerate the training of deep neural networks.

    - Faster Convergence: By keeping the distribution of layer inputs consistent, they create a smoother optimization landscape, which allows for higher learning rates and faster training.
    
    - Reduced Sensitivity to Initialization: The network becomes less dependent on the choice of initial random weights, making the training process more reliable.
    
    - Regularization: The process, especially the noise in Batch Normalization's batch-based statistics, has a slight regularizing effect that can reduce overfitting.
    
    - Combats Internal Covariate Shift: This is the core problem they solve—the phenomenon where the distribution of a layer's inputs changes as the weights of previous layers are updated during training.

## Example
Apply Batch Normalization in a TensorFlow model.

```python
import tensorflow as tf
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Dense(64, activation='relu'),
    layers.BatchNormalization(),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()
```
