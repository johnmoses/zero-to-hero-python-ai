# Dropout Regularization

## Learning Objectives
- Understand the dropout technique to prevent overfitting.
- Learn how dropout is applied during training and testing.

## Introduction
Dropout randomly deactivates neurons during training, reducing reliance on any one neuron and improving generalization.

## Core Concepts

- Dropout Rate (p)

    - Definition: A hyperparameter representing the probability that the output of any given neuron will be set to zero during a single training forward pass.

    - Typical Values: A common value is between 0.2 and 0.5. A higher rate means more aggressive regularization.
    Function: It controls the "thinning" of the network during training. A rate of 0.3 means approximately 30% of neurons in the layer will be deactivated for that training step.

- Training Behavior

    - Mechanism: For each training mini-batch, a different, randomly selected set of neurons is "dropped out" (their outputs are set to zero).

    - Effect: This forces the network to learn more robust and redundant representations. It prevents neurons from "co-adapting" or relying too heavily on the presence of a few specific other neurons. Each neuron is forced to become more independently useful.
    - Analogy: Imagine a large team working on a project. On any given day, random employees are told to take the day off. The remaining team members must learn to handle the work without them, making the entire team more resilient and less dependent on any single individual.

- Inference (Testing) Behavior

    - Mechanism: During inference (when making predictions on new data), no neurons are dropped. The full, complete network is used to maximize its predictive power.

    - Weight Scaling: To compensate for the fact that more neurons are active during inference than during training, the outputs of the layer are scaled down by a factor of (1 - p). This is known as "inverted dropout" and ensures that the expected output of each neuron remains consistent between training and testing.
    - Analogy: When the final project is due (at inference time), the entire team shows up. Because everyone was trained to be individually capable, the full team is now extremely effective and robust. The scaling is like an accounting step to average out the total effort.

## Example
Add dropout layer in a Keras model.

```python
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Dense(128, activation='relu', input_shape=(784,)),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()
```
