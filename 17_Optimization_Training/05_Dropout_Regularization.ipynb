# Dropout Regularization

## Learning Objectives
- Understand the dropout technique to prevent overfitting.
- Learn how dropout is applied during training and testing.

## Introduction
Dropout randomly deactivates neurons during training, reducing reliance on any one neuron and improving generalization.

## Core Concepts
- **Dropout Rate:** Probability of dropping a neuron.
- **Training Behavior:** Neurons dropped randomly per batch.
- **Inference Behavior:** Full network used with scaled weights.

## Example
Add dropout layer in a Keras model.

```python
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Dense(128, activation='relu', input_shape=(784,)),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()
```
