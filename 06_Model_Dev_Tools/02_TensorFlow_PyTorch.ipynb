# Introduction to TensorFlow and PyTorch

## Learning Objectives
- Understand the basics of popular deep learning frameworks TensorFlow and PyTorch.
- Learn to build simple neural networks with both.

## Introduction
TensorFlow and PyTorch are widely used frameworks for developing AI models. They offer tools to build, train, and deploy neural networks easily.

## Core Concepts
- **TensorFlow:** Graph-based computation, Keras API.
- **PyTorch:** Dynamic computation graph, eager execution.
- **Tensors:** Multi-dimensional arrays.
- **Autograd:** Automatic differentiation for training.

## Example
Create a simple neural network in TensorFlow and PyTorch.

```python
# TensorFlow example
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(5,)),
    tf.keras.layers.Dense(1)
])

model.summary()
```

```python
# PyTorch example
import torch
import torch.nn as nn

class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()
        self.fc1 = nn.Linear(5, 10)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(10, 1)
    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x

model_pt = SimpleNN()
print(model_pt)
```
