## Core Concepts
Debugging Tools

Debugging is the process of finding and fixing errors, or "bugs," in your code. In deep learning, bugs can be notoriously difficult to find because they might not cause a crash but simply lead to poor model performance.

- Print Statements: This is the most basic and universal debugging technique. By inserting print() statements (e.g., print(tensor.shape), print(tensor.dtype), print(f"Loss: {loss.item()}")) at key points in your code, you can inspect the state of your variables, the shape of your tensors, and the flow of execution. It's a quick way to verify assumptions, but it can become messy and requires you to manually add and remove the statements. It is especially useful for catching shape mismatches.

- IDE Debuggers: A much more powerful and systematic approach. An Integrated Development Environment (IDE) like VS Code or PyCharm comes with a built-in debugger (or you can use Python's pdb). A debugger allows you to:
    - Set Breakpoints: Pause the execution of your program at a specific line of code.
    - Inspect Variables: While paused, you can examine the value of every variable in your program's current scope.
    - Step Through Code: Execute your code line-by-line to watch how variables change and pinpoint the exact moment something goes wrong. This is invaluable for understanding complex code paths and diagnosing errors that are not immediately obvious.

- Logging: This is a more structured and professional alternative to print statements. A logging framework allows you to record events with different levels of severity (e.g., DEBUG, INFO, WARNING, ERROR). You can configure these logs to be saved to a file, which is essential for tracking the progress of long training runs. If a model's performance unexpectedly degrades or a crash occurs hours into training, the log file provides a detailed record of what was happening right before the issue, which is critical for post-mortem analysis.


Profiling

While debugging finds correctness issues (bugs), profiling finds performance issues (bottlenecks).

- Measuring Time (Time Profiling): This involves measuring the execution time of different parts of your code to find out what is slow. In a machine learning pipeline, the bottleneck could be in data loading, preprocessing, or the model's forward/backward pass. Tools like Python's cProfile or framework-specific profilers (e.g., TensorFlow Profiler, PyTorch Profiler) can give you a detailed breakdown of how much time is spent in each function. This is crucial for optimization, as it tells you where to focus your efforts to speed up training.

- Measuring Memory Usage (Memory Profiling): This involves tracking how much RAM (or GPU VRAM) your program is using. This is critical in deep learning, where models and datasets can be very large. A memory profiler can help you diagnose "Out of Memory" errors by showing you which objects or tensors are consuming the most memory. This can reveal memory leaks (where memory is not being released after use) or help you optimize your code by processing data in smaller batches.

Common Issues

These are classic problems that nearly every machine learning practitioner encounters.

- Shape Mismatches: This is the most common error in deep learning code. Tensor operations (like matrix multiplication) have strict rules about the dimensions (or "shapes") of the input tensors. A shape mismatch error occurs when you try to perform an operation on tensors with incompatible shapes (e.g., trying to multiply a tensor of shape [64, 100] with one of shape [10, 50]). The error message will usually tell you the shapes that were involved, and the fix is to trace your code (often with print statements or a debugger) to find where the tensor shapes became incorrect.

- Data Leaks: This is a subtle but severe logical error. Data leakage happens when information from your validation or test set accidentally "leaks" into your training process. A classic example is calculating the mean and standard deviation for data normalization using the entire dataset, and then splitting it into train/test sets. The model has now been "trained" on information from the test set, leading to overly optimistic performance metrics. The model performs well in evaluation but will fail on truly new data. The fix: Always split your data into training, validation, and test sets first, and then perform any data fitting (like normalization scalers or feature selection) only on the training data.

- Overfitting: This occurs when your model learns the training data too well, including its noise and random quirks, instead of the underlying general pattern. An overfit model will have very high accuracy on the data it was trained on, but will perform poorly on new, unseen data (like your test set). It has "memorized" the answers instead of "learning" the concept. Common techniques to combat overfitting include:
    - Regularization: Adding a penalty to the loss function for large model weights (L1/L2 regularization) or randomly deactivating neurons during training (Dropout).
    - Getting more data: A larger and more diverse training set makes it harder for the model to simply memorize everything.
    - Early Stopping: Monitoring the model's performance on the validation set during training and stopping when the validation performance stops improving, even if the training performance is still getting better.

## Example
Simple timing code and use of `%timeit` magic command in Jupyter.

```python
import time

start = time.time()
result = sum([i for i in range(1000000)])
end = time.time()

print(f"Sum: {result}, Time taken: {end - start:.4f} seconds")
```

```python
%timeit sum([i for i in range(1000000)])
```
