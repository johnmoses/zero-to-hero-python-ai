## Introduction
TensorFlow and PyTorch are widely used frameworks for developing AI models. They offer tools to build, train, and deploy neural networks easily.

## Core Concepts
TensorFlow

TensorFlow is an open-source deep learning framework developed by Google. Its design philosophy has evolved, but its core strengths are in production deployment and scalability.

- Graph-based Computation: Originally, TensorFlow operated on a "define-then-run" principle. You would first define your entire model and its computations as a static computational graph. This graph represents all the mathematical operations and their dependencies. Once defined, you would launch a session, feed data (tensors) into the graph, and run it to get the results. This approach allows for powerful optimizations and makes it easy to deploy models on various platforms (servers, mobile, web) because the graph is a self-contained, portable representation of the model.

    Note: While this static graph was TensorFlow's hallmark, TensorFlow 2.x adopted eager execution by default, making it operate more like PyTorch. However, the graph concept is still fundamental and can be explicitly used via tf.function decorators to gain performance benefits.

- Keras API: Keras is a high-level, user-friendly API for building and training neural networks. It is now the official high-level API for TensorFlow (tf.keras). Keras abstracts away much of the boilerplate code, allowing you to define complex models by simply stacking layers. It provides simple, clean methods for compiling the model (specifying the optimizer and loss function) and training it (model.fit()), making deep learning much more accessible.



PyTorch

PyTorch is an open-source deep learning framework developed by Meta AI. It is widely popular in the research community for its flexibility and intuitive, Python-like feel.

- Dynamic Computation Graph: In contrast to TensorFlow's original static graph, PyTorch uses a "define-by-run" approach. The computational graph is built dynamically, on-the-fly, as you execute operations. Every time you run your code, the graph is created from scratch.

- Eager Execution: This is a direct result of the dynamic graph. Operations are evaluated immediately, just like standard Python code. You can inspect any tensor's value at any point, use standard Python debugging tools (like pdb), and use native Python control flow (e.g., if statements, for loops) to create models that change their structure based on the input data. This makes the development and debugging process feel much more natural and straightforward.

Tensors

Tensors are the fundamental data structure in both TensorFlow and PyTorch. They are the generalization of vectors and matrices to an arbitrary number of dimensions.

- Multi-dimensional Arrays: At its core, a tensor is a container for numbers. You can think of it as a multi-dimensional array, similar to a NumPy ndarray.
    - A 0-dimensional tensor is a single number (a scalar).
    - A 1-dimensional tensor is a list of numbers (a vector).
    - A 2-dimensional tensor is a grid of numbers (a matrix). For example, a grayscale image.
    - A 3-dimensional tensor is a cube of numbers. For example, a color image (height x width x color channels) or a sequence of stock prices (stocks x timesteps x features).

- These tensors are optimized for high-performance numerical computation and can be seamlessly moved to a GPU or TPU to accelerate calculations.

Autograd (Automatic Differentiation)

Autograd is the engine that powers the training of neural networks in modern frameworks. It automates the calculation of gradients, which is essential for optimizing models.

- Automatic Differentiation for Training: Neural networks "learn" by adjusting their internal parameters (weights and biases) to minimize a "loss function," which measures how inaccurate the model's predictions are. The most common way to do this is through gradient descent, where parameters are nudged in the direction that most rapidly decreases the loss. This "direction" is determined by the gradient (the derivative) of the loss function with respect to each parameter.

- The autograd system keeps a record of every operation performed on a tensor. When you call a function like loss.backward(), it traverses this record backward (using the chain rule from calculus) to automatically and efficiently compute the gradient of the loss for every single parameter in the model. This frees the developer from having to manually derive and implement these complex derivative calculations, which is an incredibly tedious and error-prone task.

## Example
Create a simple neural network in TensorFlow and PyTorch.

```python
# TensorFlow example
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(5,)),
    tf.keras.layers.Dense(1)
])

model.summary()
```

```python
# PyTorch example
import torch
import torch.nn as nn

class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()
        self.fc1 = nn.Linear(5, 10)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(10, 1)
    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x

model_pt = SimpleNN()
print(model_pt)
```
