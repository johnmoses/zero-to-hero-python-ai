## Introduction
Experiment tracking enables reproducibility and helps compare different model versions, hyperparameters, and results efficiently.

## Core Concepts
Logging Parameters

This is the practice of systematically recording the configuration used for each training run. In machine learning, these configuration settings are called hyperparameters.

- What it is: For every experiment you run, you save the exact values of all the hyperparameters that define it. This includes:
    - Model Hyperparameters: Number of layers, number of neurons per layer, activation functions, dropout rate, etc.
    - Training Hyperparameters: Learning rate, batch size, number of epochs, optimizer choice (e.g., Adam, SGD), etc.
    - Data Hyperparameters: The version of the dataset used, feature engineering steps applied, etc.
    
- Why it's important: Finding the right combination of hyperparameters is often an empirical process of trial and error. If you don't log them, you can easily lose track of which combination led to your best-performing model. Logging parameters ensures reproducibility and allows you to systematically analyze how different settings affect the outcome, turning random guesses into a structured search.


Tracking Metrics

While parameters are the inputs to your experiment, metrics are the outputs that measure its performance.

- What it is: This involves monitoring and recording key performance indicators during and after training. This includes:
    - Performance Metrics: Loss, accuracy, precision, recall, F1-score, Mean Squared Error (MSE), etc., for both the training and validation sets.
    - System Metrics: Training time (runtime), GPU/CPU utilization, memory usage.

- Why it's important: Tracking metrics over time (e.g., epoch by epoch) is essential for diagnosing the training process. By plotting the training and validation loss, you can instantly see if your model is learning effectively or if it's overfitting (i.e., validation loss starts increasing while training loss continues to decrease). Comparing final metrics across different experiments is how you determine which one was most successful.

Version Control

This concept bridges the gap between your code and your experimental results, ensuring true end-to-end reproducibility.

- What it is: It's the practice of linking every experiment run to the exact version of the code that produced it. In practice, this usually means automatically recording the Git commit hash along with the logged parameters and metrics for a given run.

- Why it's important: Your code is constantly evolving. A change in your data preprocessing logic or model architecture can have a huge impact on performance. If you discover that an experiment from two months ago was actually the best one, you need a reliable way to get back to the exact state of the project at that time. By linking a run to a Git hash, you can check out the precise code version, use the logged hyperparameters, and reproduce the result perfectly. It eliminates the "it worked on my machine last week" problem.

Tools

Manually managing all of the above in spreadsheets is tedious and error-prone. Specialized tools automate this process.

- MLflow: An open-source platform for managing the end-to-end machine learning lifecycle. It's a popular choice for teams that want to host their own tracking server. It provides components for:
    - MLflow Tracking: An API and UI for logging parameters, metrics, and artifacts (like model files).
    - MLflow Projects: A format for packaging your code to ensure it can be re-run reproducibly.
    - MLflow Models: A convention for packaging models that allows them to be deployed in a variety of environments.

- Weights & Biases (W&B): A commercial service (with a generous free tier) that provides a polished, hosted platform for experiment tracking. It is known for its ease of use and powerful visualization tools. With just a few lines of code, it can automatically log metrics, parameters, and system stats (like GPU temperature and usage), and present them in a beautiful, interactive web dashboard where you can easily compare dozens of experiments.

- TensorBoard: A visualization toolkit that comes packaged with TensorFlow (but also works well with PyTorch). You write special log files from your script, and TensorBoard turns them into an interactive web application for visualizing metrics, model graphs, histograms of weights, and more. It is excellent for debugging and monitoring a single training run in detail, but it is less focused on the high-level management and comparison of many different experiments, where tools like MLflow and W&B excel.

## Example
Basic MLflow usage (requires installation).

```python
# !pip install mlflow
import mlflow
import random

with mlflow.start_run():
    mlflow.log_param("learning_rate", 0.01)
    accuracy = random.uniform(0.8, 0.9)
    mlflow.log_metric("accuracy", accuracy)
    print(f"Logged accuracy: {accuracy}")
```
