## Introduction
Debugging and profiling help improve code correctness and efficiency in AI workflows, essential for complex pipelines and models.

## Core Concepts
- **Debugging Tools:** Print statements, IDE debuggers, logging.
- **Profiling:** Measuring time and memory usage.
- **Common Issues:** Shape mismatches, data leaks, overfitting.

## Example
Simple timing code and use of `%timeit` magic command in Jupyter.

```python
import time

start = time.time()
result = sum([i for i in range(1000000)])
end = time.time()

print(f"Sum: {result}, Time taken: {end - start:.4f} seconds")
```

```python
%timeit sum([i for i in range(1000000)])
```
