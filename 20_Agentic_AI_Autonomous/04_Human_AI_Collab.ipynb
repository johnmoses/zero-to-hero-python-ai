## Introduction
Human-AI collaboration enhances decision-making by combining human judgment with AI capabilities.

## Key Concepts
- **Explainable AI:** Making AI decisions interpretable.
- **Shared control:** Balancing authority between human and AI.
- **Trust calibration:** Ensuring appropriate reliance on AI.
- **User experience design:** Creating effective AI interfaces.
