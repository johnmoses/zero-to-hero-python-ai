## Introduction
Agent frameworks enable AI systems to perceive, reason, and act autonomously in complex environments.

## Core Architectures
- **Reactive Agents:** Respond directly to environment stimuli.
- **Deliberative Agents:** Use internal models to plan.
- **Hybrid Agents:** Combine reactive and deliberative approaches.
- **BDI Model:** Belief-Desire-Intention framework.

## Example
Simple rule-based reactive agent pseudocode.

```python
def reactive_agent(percept):
    if percept == 'obstacle':
        action = 'turn'
    else:
        action = 'move_forward'
    return action

print(reactive_agent('clear'))
print(reactive_agent('obstacle'))
```
