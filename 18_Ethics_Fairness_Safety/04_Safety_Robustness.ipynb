## Introduction

## Core Concepts
- **Adversarial Examples:** Inputs designed to fool AI models.
- **Robustness Testing:** Evaluating model behavior under perturbations.
- **Defenses:** Adversarial training, input sanitization, detection methods.

## Example
Simple adversarial perturbation generation on image data (conceptual).

```python
import numpy as np

def fgsm_attack(image, epsilon, gradient):
    # Fast Gradient Sign Method
    perturbed_image = image + epsilon * np.sign(gradient)
    return np.clip(perturbed_image, 0, 1)

# Example usage with dummy data
image = np.array([[0.5, 0.5], [0.5, 0.5]])
gradient = np.array([[0.1, -0.2], [0.3, -0.4]])
epsilon = 0.1
perturbed = fgsm_attack(image, epsilon, gradient)
print("Original image:\n", image)
print("Perturbed image:\n", perturbed)
```
