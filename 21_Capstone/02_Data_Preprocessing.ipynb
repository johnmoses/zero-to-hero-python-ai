## Introduction
Effective preprocessing directly affects model performance and generalizability.

## Common Techniques

- Handling missing values: Imputing (e.g., with mean/median) or removing empty data points to allow for model training.

- Encoding categorical variables: Converting non-numeric data (e.g., 'red', 'green') into numbers so algorithms can process them.

- Normalization and scaling: Adjusting the range of numeric features to a common scale (e.g., 0 to 1) to prevent any single feature from dominating.

- Feature extraction and selection: Creating new features from existing data or choosing the most impactful ones to improve model performance.


```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

# Sample data
data = pd.DataFrame({
    'age': [25, 30, None, 35],
    'gender': ['M', 'F', 'F', None],
    'income': [50000, 60000, 55000, 65000]
})

# Impute missing values
imputer = SimpleImputer(strategy='mean')
data['age'] = imputer.fit_transform(data[['age']])

# Encode categorical
encoder = OneHotEncoder(sparse=False, handle_unknown='ignore')
gender_encoded = encoder.fit_transform(data[['gender']].fillna('Unknown'))

# Normalize income
scaler = StandardScaler()
data['income_scaled'] = scaler.fit_transform(data[['income']])

print(data)
print("Gender Encoded:\n", gender_encoded)
```
