## Introduction
Evaluating model performance and interpreting results ensures actionable and trustworthy insights.

## Common Metrics

- Accuracy, precision, recall, F1: Standard scores for classification tasks measuring overall correctness, positive prediction quality, ability to find all positives, and a balance between the two, respectively.

- MSE, MAE, R²: Core metrics for regression tasks measuring average squared error, average absolute error, and the proportion of variance explained by the model.

- Confusion Matrices and ROC curves: Visual tools to assess a classifier's performance, with the matrix showing raw prediction counts (TP/FP/TN/FN) and the ROC curve illustrating performance across different thresholds.

## Interpretation Techniques

- Feature importance: A method to score and rank which input features had the most influence on the model's predictions.

- SHAP and LIME explanations: Techniques that explain a model's output for a single prediction by showing how much each feature contributed to that specific outcome.

## Example
Display classification report using scikit-learn.

```python
from sklearn.metrics import classification_report
import numpy as np

y_true = np.array([0, 1, 1, 0, 1])
y_pred = np.array([0, 1, 0, 0, 1])
report = classification_report(y_true, y_pred)
print(report)
```
